````python
# ---------------------------------------------------------------------
# Generación del registro trimestral de consulta externa
# Datos SINTÉTICOS con semilla fija: tus resultados serán idénticos a los del ejemplo.
#
# Se guardan como cuatro LISTAS PARALELAS: el elemento i de cada una describe
# la misma cita. Es la estructura más ingenua posible y la usaremos a propósito,
# porque sus problemas son el mejor argumento para lo que viene después.
# ---------------------------------------------------------------------
import random

ESPECIALIDADES = ["Medicina interna", "Pediatría", "Ortopedia", "Ginecología"]
DIAS = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]
N_CITAS = 900


def generar_registro(semilla=SEMILLA, n=N_CITAS):
    '''Genera el registro trimestral sintético de consulta externa.

    Devuelve cuatro listas paralelas: esperas, especialidades, horas y días.
    Que sea una función y no un bloque suelto es deliberado: más adelante
    destruiremos los datos a propósito y necesitarás reconstruirlos.
    '''
    rng = random.Random(semilla)
    esperas, especialidades, horas, dias = [], [], [], []

    for _ in range(n):
        hora = rng.randrange(7, 17)
        especialidad = rng.choice(ESPECIALIDADES)
        dia = rng.choice(DIAS)

        # La agenda acumula retraso a lo largo de la jornada.
        base = 12 + (hora - 7) * 3.1

        # Cada especialidad arrastra su propio recargo estructural.
        if especialidad == "Ortopedia":
            base += 14
        elif especialidad == "Pediatría":
            base += 6
        elif especialidad == "Ginecología":
            base += 3

        if dia == "Lunes":
            base += 9          # el lunes arrastra lo represado del fin de semana

        espera = base + rng.gauss(0, 7)

        # Un 8 % de las citas sufre una interrupción del servicio.
        if rng.random() < 0.08:
            espera += rng.uniform(40, 120)

        esperas.append(round(max(espera, 1.0), 1))   # nadie espera menos de un minuto
        especialidades.append(especialidad)
        horas.append(hora)
        dias.append(dia)

    return esperas, especialidades, horas, dias


esperas, especialidades, horas, dias = generar_registro()

print("Citas registradas:", len(esperas))
print("Primeras cinco esperas:", esperas[:5])
print("Primeras cinco especialidades:", especialidades[:5])
print("¿Las cuatro listas tienen el mismo largo?",
      len(esperas) == len(especialidades) == len(horas) == len(dias))
````

### 1.1 El acumulador ya no alcanza

En el Módulo 1 calculabas el promedio así: una variable en cero, un bucle que sumaba, una división al final. Es una solución elegante y tiene una propiedad valiosa que conviene no despreciar, porque el resto del módulo la perderá: **consume una cantidad de memoria que no depende del número de citas**. Con mil o con mil millones, el acumulador ocupa lo mismo.

Su límite es igual de claro. Ejecuta la celda siguiente y observa qué puedes responder y qué no.

````python
# Enfoque del Módulo 1: acumular sin guardar.
total = 0
n = 0
for espera in esperas:          # recorremos, pero imagina que el dato llega y se pierde
    total += espera
    n += 1

print(f"Espera media : {total / n:.1f} min sobre {n} citas")
print()
print("Con solo el acumulador NO puedes responder:")
print("  · ¿cuál fue la espera más larga?")
print("  · ¿cuál es la mediana?")
print("  · ¿qué porcentaje de pacientes esperó más de una hora?")
print("  · ¿cómo se ve la distribución?")
````

Ninguna de esas cuatro preguntas es exótica. La segunda y la cuarta son, de hecho, las que hacen falta para saber si el promedio de 42 minutos describe a alguien real o es un artefacto estadístico.

Una **lista** guarda las observaciones y las deja disponibles para volver a mirarlas tantas veces como haga falta.

````python
print("Espera máxima :", max(esperas), "min")
print("Espera mínima :", min(esperas), "min")
print("Número de citas:", len(esperas))

ordenadas = sorted(esperas)
mitad = len(ordenadas) // 2
mediana = ordenadas[mitad] if len(ordenadas) % 2 == 1 else (ordenadas[mitad - 1] + ordenadas[mitad]) / 2

print(f"Media    : {sum(esperas) / len(esperas):.1f} min")
print(f"Mediana  : {mediana:.1f} min")
````

Ahí está el primer hallazgo del módulo, y no es menor: **la media supera a la mediana**. Eso significa que la distribución tiene una cola hacia la derecha, es decir, un grupo de pacientes cuya espera es mucho mayor que la del resto y que arrastra el promedio hacia arriba. La mitad de las personas espera menos que la mediana, de modo que el número que la gerencia viene discutiendo describe a una minoría, no al paciente típico.

🌱 **Analogía.** Un acumulador es como llevar la cuenta mental del gasto del mes: sabes cuánto llevas, y no puedes revisar en qué se fue. Una lista es guardar todos los recibos: ocupa espacio en un cajón y a cambio permite responder preguntas que ni siquiera te habías planteado cuando empezaste a guardarlos.

🔬 **Nota técnica.** Una lista de Python es un arreglo dinámico de **referencias a objetos**, no de valores. Cada elemento es un puntero, de modo que la lista puede ser heterogénea y cada acceso implica una indirección. Al crecer, `append` amortiza el costo reservando más espacio del necesario, lo que da un costo promedio constante por inserción aunque alguna operación individual sea cara. Esa flexibilidad se paga en memoria y en velocidad, y es exactamente lo que NumPy elimina en la sección 11 al exigir un solo tipo y almacenar los valores contiguos.

### 1.2 Crear y hacer crecer una lista

````python
# Cuatro formas de crear listas
vacia = []
literal = [35, 48, 62, 27, 41]
repetida = [0] * 5                      # cinco ceros
desde_rango = list(range(7, 17))        # las horas de la jornada

print("Vacía        :", vacia, "· longitud", len(vacia))
print("Literal      :", literal)
print("Repetida     :", repetida)
print("Desde rango  :", desde_rango)

# Crecer: append añade UN elemento; extend añade todos los de otra secuencia
turnos = []
turnos.append("07:00")
turnos.append("07:20")
turnos.extend(["07:40", "08:00"])
print("Turnos       :", turnos, "· longitud", len(turnos))
````

⚠️ **Error frecuente.** Confundir `append` con `extend`. `turnos.append(["07:40", "08:00"])` no añade dos elementos: añade **uno solo**, que resulta ser una lista. La longitud crece en uno y el contenido queda anidado, con la particularidad de que no se produce ningún error y el problema aparece más tarde, cuando algo espera una cadena de texto y recibe una lista.

### 1.3 La trampa que define esta sección: alias frente a copia

Esta es, con diferencia, la fuente de errores silenciosos más costosa de todo el módulo. Léela dos veces.

````python
registro_original = [35, 48, 62, 27, 41]

copia_aparente = registro_original          # ← esto NO copia nada
copia_aparente.append(999)

print("Lo que modifiqué :", copia_aparente)
print("El original      :", registro_original)
print("¿Son el mismo objeto?", registro_original is copia_aparente)
````

El original cambió, aunque nadie lo tocó. La asignación con `=` **no copia la lista**: le pone un segundo nombre al mismo objeto. Cualquier modificación hecha a través de un nombre es visible desde el otro, porque no hay dos listas, hay una.

🧩 **Trazado de estado.**

| Instrucción | Objetos en memoria | Nombres que apuntan a cada uno |
|---|---|---|
| `registro_original = [35, 48, ...]` | Lista A, con 5 elementos | `registro_original` |
| `copia_aparente = registro_original` | Lista A, con 5 elementos | `registro_original`, `copia_aparente` |
| `copia_aparente.append(999)` | Lista A, con 6 elementos | `registro_original`, `copia_aparente` |

Para copiar de verdad hay tres formas equivalentes.

````python
original = [35, 48, 62, 27, 41]

copia_1 = original.copy()      # la más explícita, y la recomendada
copia_2 = original[:]          # segmentación completa: idiomática pero menos legible
copia_3 = list(original)       # constructor

copia_1.append(999)

print("Original tras modificar la copia:", original)
print("¿copia_1 es el mismo objeto?", copia_1 is original)
print("¿copia_2 tiene el mismo contenido?", copia_2 == original)
````

Fíjate en la distinción entre los dos operadores de la última línea. `is` pregunta si son **el mismo objeto**; `==` pregunta si tienen **el mismo contenido**. Dos listas distintas con idénticos elementos son iguales con `==` y no lo son con `is`.

🔬 **Nota técnica.** Los tres métodos anteriores producen una copia **superficial** (*shallow copy*): duplican la lista externa, pero sus elementos siguen siendo referencias a los mismos objetos. Con números y cadenas eso es indiferente, porque son inmutables. Con listas anidadas, que aparecen en la sección 5, no lo es en absoluto: modificar una sublista de la copia también modifica la del original. La solución es `copy.deepcopy`, que reconstruye la estructura completa a cambio de un costo notablemente mayor.

### 🎯 Ejercicio rápido 1

¿Por qué `lista_b = lista_a` no sirve para hacer una copia de seguridad antes de una transformación arriesgada?

### 2.2 Segmentación: extraer un tramo

````python
print("Las cinco primeras   :", esperas[:5])
print("Las cinco últimas    :", esperas[-5:])
print("De la 10 a la 15     :", esperas[10:15])
print("Una de cada cien     :", esperas[::100])
print("El registro invertido, tres primeros:", esperas[::-1][:3])
````

La sintaxis es `lista[inicio:fin:paso]`. El **inicio se incluye y el fin se excluye**, exactamente igual que en `range`, de modo que `esperas[10:15]` devuelve cinco elementos: del 10 al 14. Omitir un extremo significa "desde el principio" o "hasta el final", y un paso negativo recorre en sentido inverso.

🌱 **Analogía.** Piensa en las marcas de una regla y no en los objetos que mide. `[10:15]` no señala los elementos décimo a decimoquinto, sino el tramo comprendido **entre la marca 10 y la marca 15**. Con esa imagen, que el fin quede excluido deja de ser una arbitrariedad: es la única forma de que dos tramos consecutivos, `[0:5]` y `[5:10]`, no compartan ningún elemento ni dejen ninguno fuera.

🔬 **Nota técnica.** La segmentación de una lista devuelve siempre una **lista nueva**, con una copia superficial de las referencias, de modo que `nueva = original[10:20]` no comparte identidad con el original y modificarla no lo afecta. Eso la distingue de NumPy, donde la segmentación de un `ndarray` devuelve una **vista** que sí comparte memoria con el arreglo original: escribir en la vista modifica el arreglo. Esa diferencia, que aparecerá en la sección 12, es una de las confusiones más frecuentes al pasar de listas a arreglos.

### 3.1 Recorrer con y sin índice

````python
# Sin índice: cuando solo importan los valores
suma = 0
for espera in esperas:
    suma += espera
print(f"Media por recorrido directo : {suma / len(esperas):.1f} min")

# Con índice: cuando hace falta cruzar listas paralelas
criticas_ortopedia = 0
for i in range(len(esperas)):
    if esperas[i] > 60 and especialidades[i] == "Ortopedia":
        criticas_ortopedia += 1
print("Citas de Ortopedia con espera > 60 min :", criticas_ortopedia)

# enumerate: valores e índice a la vez, que es lo idiomático
peor_espera = 0
peor_indice = 0
for i, espera in enumerate(esperas):
    if espera > peor_espera:
        peor_espera = espera
        peor_indice = i

print(f"Peor espera: {peor_espera} min · {especialidades[peor_indice]} · "
      f"{dias[peor_indice]} a las {horas[peor_indice]}:00")
````

⚠️ **Error frecuente.** Escribir `for i in range(len(lista))` cuando en realidad solo se necesitan los valores. Funciona, y es más largo, más lento y más fácil de equivocar. La regla práctica: si dentro del bucle nunca escribes `lista[i]`, no necesitabas el índice; si lo escribes para cruzar varias listas, `enumerate` o `zip` lo expresan mejor.

### 3.2 Transformar: en el sitio o creando una lista nueva

Esta distinción vuelve a aparecer, ahora con consecuencias distintas.

````python
# Transformación creando una lista NUEVA: el original queda intacto
esperas_horas = []
for espera in esperas:
    esperas_horas.append(round(espera / 60, 2))

print("Original, primeros tres :", esperas[:3])
print("En horas, primeros tres :", esperas_horas[:3])
print("¿El original cambió?", len(esperas), "citas siguen ahí")
````

🔬 **Nota técnica.** La alternativa, modificar en el sitio con `esperas[i] = esperas[i] / 60`, ahorra memoria y **destruye el dato original de forma irreversible**. En análisis exploratorio esa economía casi nunca compensa: si más adelante descubres que la conversión estaba mal, o que necesitabas los minutos, tendrías que volver a generar o a leer los datos. La convención profesional es transformar creando objetos nuevos mientras el volumen lo permita, y reservar la modificación en el sitio para los casos en que la memoria sea realmente el cuello de botella.

### 3.3 Ordenar: `sorted` frente a `sort`

````python
prueba = [35, 48, 62, 27, 41]

nueva = sorted(prueba)                 # devuelve una lista NUEVA
print("sorted → nueva  :", nueva)
print("sorted → original:", prueba, "· intacto")

prueba.sort()                          # modifica EN EL SITIO y devuelve None
print("sort   → original:", prueba, "· modificado")
print("¿Qué devuelve sort()?", prueba.sort())
````

⚠️ **Error frecuente y silencioso.** Escribir `lista = lista.sort()`. Como `sort` devuelve `None`, esa línea reemplaza la lista por `None` y todo lo que venga después falla con un mensaje confuso, del tipo `TypeError: object of type 'NoneType' has no len()`, varias líneas más abajo del punto donde estaba el problema. La regla es sencilla: `sorted` para asignar, `sort` para modificar sin asignar.

### 3.4 Ordenar por un criterio propio

El parámetro `key` recibe una función que decide, para cada elemento, el valor por el que se debe comparar.

````python
# Ordenar las citas por espera, conservando su índice original.
# zip empareja elemento a elemento; lo veremos a fondo en la sección 7.
indices = list(range(len(esperas)))
por_espera = sorted(indices, key=lambda i: esperas[i], reverse=True)

print("Las cinco esperas más largas del trimestre:")
for posicion, i in enumerate(por_espera[:5], start=1):
    print(f"  {posicion}. {esperas[i]:>6.1f} min · {especialidades[i]:<17} · "
          f"{dias[i]:<10} · {horas[i]}:00")
````

La estrategia de ordenar **los índices** en lugar de los valores es la respuesta al problema de las listas paralelas: conservando el índice, se puede recuperar toda la información de cada cita. Es una solución correcta y también un síntoma de que la estructura de datos elegida no es la adecuada, cosa que resolveremos en la sección 5 con listas bidimensionales y en la 8 con diccionarios.

🔬 **Nota técnica.** El ordenamiento de Python es **estable**: los elementos que empatan según la clave conservan entre sí el orden que tenían antes. Esa garantía, que muchos lenguajes no ofrecen, permite ordenar por criterios sucesivos aplicando el menos importante primero, sin necesidad de una clave compuesta. El algoritmo se llama Timsort, fue diseñado por Tim Peters en 2002, y aprovecha los tramos ya ordenados que suelen aparecer en los datos reales; su costo es del orden de *n* log *n* en el peor caso y lineal cuando la lista ya está casi ordenada.

### 4.1 La tupla: una secuencia que no puede cambiar

Una tupla se escribe con paréntesis y se comporta como una lista en todo salvo en una cosa: **no se puede modificar después de creada**.

````python
cita = (73.4, "Ortopedia", 14, "Lunes")

print("La tupla completa :", cita)
print("Espera            :", cita[0], "min")
print("Especialidad      :", cita[1])
print("Longitud          :", len(cita))
print("Se recorre igual  :", [campo for campo in cita][:2], "...")
````

````python
cita = (73.4, "Ortopedia", 14, "Lunes")
cita[0] = 80.0
````

El error es `TypeError: 'tuple' object does not support item assignment`. No es una carencia: es una **garantía**. Una tupla que llega a tus manos no cambiará mientras la uses, aunque la pases a una función escrita por otra persona.

🌱 **Analogía.** Una lista es una pizarra: cualquiera que la tenga a la vista puede borrar y reescribir. Una tupla es un documento impreso y firmado: para cambiar algo hay que emitir uno nuevo, y el anterior sigue siendo evidencia de lo que decía antes. En un registro clínico esa diferencia no es estilística, es la base de la trazabilidad.

🔬 **Nota técnica.** La inmutabilidad tiene una consecuencia técnica que aparecerá en la sección 8: las tuplas son **dispersables** (*hashable*) y las listas no, de modo que una tupla puede ser clave de un diccionario o elemento de un conjunto y una lista no puede. La razón es que la dispersión debe producir siempre el mismo valor para el mismo objeto, y eso es imposible de garantizar si el objeto puede cambiar. Conviene matizar que la inmutabilidad de la tupla es **superficial**: `([1, 2], 3)` es una tupla que contiene una lista, y esa lista sí se puede modificar, lo que además la vuelve no dispersable.

### 4.2 Desempaquetado: la sintaxis que más se usa en la práctica

````python
cita = (73.4, "Ortopedia", 14, "Lunes")

espera, especialidad, hora, dia = cita        # desempaquetado
print(f"{especialidad} · {dia} a las {hora}:00 · {espera} min")

# Intercambiar dos variables sin auxiliar: es desempaquetado disfrazado
a, b = 1, 2
a, b = b, a
print("Intercambio:", a, b)

# El guion bajo señala un campo que no se va a usar
espera, _, hora, _ = cita
print("Solo espera y hora:", espera, hora)
````

⚠️ **Error frecuente.** Desempaquetar con un número de nombres distinto al de elementos. `a, b = (1, 2, 3)` lanza `ValueError: too many values to unpack`, y es un error ruidoso y por tanto benigno. El caso peligroso es el contrario: si el sistema origen añade una columna al registro, el desempaquetado empieza a fallar en todas partes a la vez, lo que es incómodo y mucho mejor que asignar en silencio los valores corridos.

### 4.3 Retorno múltiple: donde la tupla aparece sin que la escribas

````python
def describir(valores):
    '''Devuelve tamaño, mínimo, máximo y media de una lista de números.'''
    n = len(valores)
    return n, min(valores), max(valores), sum(valores) / n


resultado = describir(esperas)
print("Lo que devuelve la función :", type(resultado).__name__)
print("Contenido                  :", tuple(round(v, 1) if isinstance(v, float) else v for v in resultado))

n, minimo, maximo, media = describir(esperas)
print(f"\n{n} citas · mínimo {minimo} · máximo {maximo} · media {media:.1f} min")
````

Cuando una función devuelve varios valores separados por comas, Python los empaqueta en una tupla. No hay sintaxis especial: `return a, b, c` construye la tupla `(a, b, c)`, y quien llama la desempaqueta.

### 4.4 Cada cita en un solo objeto

Aquí resolvemos, por fin, el problema que las secciones anteriores dejaron abierto.

````python
# Agrupamos las cuatro listas paralelas en UNA lista de tuplas.
citas = []
for i in range(len(esperas)):
    citas.append((esperas[i], especialidades[i], horas[i], dias[i]))

print("Total de citas :", len(citas))
print("Primera cita   :", citas[0])
print("Última cita    :", citas[-1])

# Y ahora ordenar ya no puede desalinear nada: se mueve la cita completa.
peores = sorted(citas, reverse=True)          # ordena por el primer campo: la espera

print("\nLas tres peores esperas del trimestre:")
for espera, especialidad, hora, dia in peores[:3]:
    print(f"  {espera:>6.1f} min · {especialidad:<17} · {dia:<10} · {hora}:00")
````

Compara este resultado con el error silencioso 2 de la sección 3. Allí la peor espera se atribuía a la especialidad equivocada; aquí eso **ya no puede ocurrir**, porque no hay dos estructuras que puedan desalinearse: hay una sola, y cada elemento contiene la cita completa.

Ese es el principio general que conviene llevarse del módulo entero: **la mejor defensa contra un error no es recordarlo, es elegir una estructura de datos en la que ese error sea imposible de cometer**.

🔬 **Nota técnica.** El orden de los campos dentro de la tupla deja de ser arbitrario en cuanto se ordena: `sorted` compara tuplas lexicográficamente, campo por campo, de modo que el primer campo domina y los siguientes solo deciden los empates. Poner la espera en primer lugar hizo que `sorted(citas)` ordenara por espera sin necesidad de `key`. Esa comodidad es también una fragilidad, porque el comportamiento depende de un orden que nadie declaró; con muchos campos conviene ser explícito con `key=lambda c: c[0]` aunque resulte más largo.

### 5.1 Una tabla es una lista de filas

````python
# Cada fila es una cita; cada columna, un campo.
tabla = []
for espera, especialidad, hora, dia in citas:
    tabla.append([espera, especialidad, hora, dia])

ENCABEZADOS = ["espera_min", "especialidad", "hora", "dia"]

print("Filas    :", len(tabla))
print("Columnas :", len(tabla[0]))
print("\nPrimeras tres filas:")
for fila in tabla[:3]:
    print("  ", fila)

print("\nAcceso por fila y columna:")
print("  tabla[0][0] =", tabla[0][0], "· la espera de la primera cita")
print("  tabla[0][1] =", tabla[0][1], "· su especialidad")
````

### 5.2 Recorrido anidado y agregación por columna

````python
COL_ESPERA, COL_ESPECIALIDAD, COL_HORA, COL_DIA = 0, 1, 2, 3   # nombres para los índices

suma = 0
maxima = 0
for fila in tabla:
    suma += fila[COL_ESPERA]
    if fila[COL_ESPERA] > maxima:
        maxima = fila[COL_ESPERA]

print(f"Media   : {suma / len(tabla):.1f} min")
print(f"Máxima  : {maxima} min")

# Extraer una columna completa
columna_horas = []
for fila in tabla:
    columna_horas.append(fila[COL_HORA])
print("Horas distintas presentes:", sorted(set(columna_horas)))
````

Dar nombre a los índices con constantes es una defensa barata contra un error caro. `fila[COL_ESPERA]` dice qué se está leyendo; `fila[0]` obliga a recordarlo, y quien lea el código dentro de seis meses no lo recordará.

### 5.3 Una matriz de verdad: especialidades por hora

````python
HORAS = sorted(set(horas))                    # 7 a 16
ESPEC = sorted(set(especialidades))           # las cuatro especialidades

# Matriz de conteos: una fila por especialidad, una columna por hora.
conteo = []
for _ in ESPEC:
    conteo.append([0] * len(HORAS))           # ojo con esta línea: la sección 5.4 explica por qué

for espera, especialidad, hora, dia in citas:
    f = ESPEC.index(especialidad)
    c = HORAS.index(hora)
    conteo[f][c] += 1

print("Citas por especialidad y hora")
print(" " * 18 + "".join(f"{h:>5}" for h in HORAS))
for f, especialidad in enumerate(ESPEC):
    fila_txt = "".join(f"{v:>5}" for v in conteo[f])
    print(f"{especialidad:<18}{fila_txt}   total {sum(conteo[f])}")

totales_hora = []
for c in range(len(HORAS)):
    totales_hora.append(sum(conteo[f][c] for f in range(len(ESPEC))))
print(f"{'TOTAL':<18}" + "".join(f"{v:>5}" for v in totales_hora))
````

### 5.4 🕵️ Error silencioso 3 · la multiplicación que comparte filas

La celda siguiente **no produce ningún error** y su resultado es absurdo. Ejecútala.

````python
matriz = [[0] * 3] * 4      # ¿cuatro filas de tres ceros?

print("Aspecto inicial:")
for fila in matriz:
    print("  ", fila)

matriz[0][0] = 99           # modificamos solo la primera fila

print("\nDespués de escribir un 99 en la fila 0:")
for fila in matriz:
    print("  ", fila)
print("\n¿Las cuatro filas son el mismo objeto?", matriz[0] is matriz[1] is matriz[2])
````

<details><summary><b>Ver diagnóstico y solución</b></summary>

El 99 aparece en las cuatro filas. `[[0] * 3] * 4` no crea cuatro filas: crea **una** fila y repite cuatro veces la misma referencia. Los cuatro elementos de la lista externa apuntan al mismo objeto, de modo que escribir en cualquiera de ellos escribe en todos.

Es el error de alias de la sección 1 en su forma más difícil de ver, porque la sintaxis parece describir cuatro cosas distintas. La solución correcta es construir cada fila por separado:

```
matriz = []
for _ in range(4):
    matriz.append([0] * 3)          # una lista nueva en cada vuelta
```

o, con la comprensión que verás en la sección 7, `matriz = [[0] * 3 for _ in range(4)]`.

Repara en que la línea marcada de la celda de la sección 5.3 usa precisamente la construcción correcta: `conteo.append([0] * len(HORAS))` dentro de un bucle. Si hubiera usado la multiplicación externa, la matriz de conteos habría acumulado todas las especialidades en una sola fila repetida, y el resultado habría parecido plausible.

<details><summary><b>Ver solución</b></summary>

Porque la multiplicación repite referencias en ambos casos, y lo que cambia es la naturaleza de lo repetido. ## 6. Pilas y colas · Stacks and queues

**Al terminar esta sección podrás:** usar una lista como pila, explicar por qué usarla como cola es una mala idea, emplear `deque` para colas eficientes, y reconocer que una política de atención es una decisión de diseño con consecuencias sobre personas.

### 6.1 La pila: sale primero lo último que entró

````python
# Una pila (LIFO) se implementa con append y pop.
pila_revisiones = []

pila_revisiones.append("Revisar agenda de Ortopedia")
pila_revisiones.append("Verificar registro del lunes")
pila_revisiones.append("Atender consulta urgente")

print("Pendientes:", pila_revisiones)
print("Atiendo    :", pila_revisiones.pop())      # sale el ÚLTIMO que entró
print("Atiendo    :", pila_revisiones.pop())
print("Quedan     :", pila_revisiones)
````

`append` apila y `pop` sin argumentos desapila desde el final. Ambas operaciones cuestan **tiempo constante**, porque el final de una lista es el extremo barato.

🌱 **Analogía.** Una pila de bandejas en una cafetería: se pone arriba y se toma de arriba. La bandeja del fondo puede quedarse ahí para siempre, cosa que resulta irrelevante con bandejas y muy relevante con personas.

### 6.2 La cola: sale primero el que llegó primero

````python
from collections import deque
import time

# Una cola (FIFO) con lista: pop(0) funciona y es caro.
cola_lista = list(range(200_000))
inicio = time.perf_counter()
while cola_lista:
    cola_lista.pop(0)
costo_lista = time.perf_counter() - inicio

# La misma cola con deque
cola_deque = deque(range(200_000))
inicio = time.perf_counter()
while cola_deque:
    cola_deque.popleft()
costo_deque = time.perf_counter() - inicio

print(f"200 000 extracciones con lista.pop(0) : {costo_lista:>7.3f} s")
print(f"200 000 extracciones con deque.popleft: {costo_deque:>7.3f} s")
print(f"La lista tardó {costo_lista / costo_deque:.0f} veces más")
````

🔬 **Nota técnica.** La diferencia no es de constantes, es de **orden de complejidad**. Una lista de Python almacena sus elementos de forma contigua, de modo que extraer el primero obliga a desplazar todos los demás una posición: el costo de una extracción es proporcional al tamaño de la cola, y vaciarla entera cuesta del orden de *n* al cuadrado. Un `deque` está implementado como una lista doblemente enlazada de bloques y permite insertar y extraer por ambos extremos en tiempo constante, de modo que vaciarla cuesta del orden de *n*. Con doscientos mil elementos la diferencia ya es visible; con dos millones, la versión con lista deja de ser viable.

### 6.3 Simular la sala de espera

````python
from collections import deque

# Los primeros veinte pacientes de la jornada, en orden de llegada.
sala = deque()
for i in range(20):
    espera, especialidad, hora, dia = citas[i]
    sala.append((i, especialidad, espera))

print("Pacientes en sala:", len(sala))

# Atención por orden de llegada (FIFO)
atendidos_fifo = []
while sala:
    atendidos_fifo.append(sala.popleft())

print("\nOrden de atención FIFO, primeros cinco:")
for i, especialidad, espera in atendidos_fifo[:5]:
    print(f"  paciente {i:>2} · {especialidad:<17} · esperó {espera:>6.1f} min")

print(f"\nEspera media con FIFO: {sum(e for _, _, e in atendidos_fifo) / len(atendidos_fifo):.1f} min")
````

### 6.4 Cambiar la política cambia a quién le va bien

````python
# Política alternativa: atender primero a quien menos va a esperar (SJF).
# Es la política que MEJOR promedio produce, y no la más justa.
lote = [(i, especialidades[i], esperas[i]) for i in range(20)]

orden_llegada = list(lote)
orden_corta_primero = sorted(lote, key=lambda t: t[2])

def espera_acumulada(orden):
    '''Suma el tiempo que cada paciente pasa esperando su turno en la cola.'''
    acumulado = 0
    total = 0
    for _, _, servicio in orden:
        total += acumulado          # lo que este paciente esperó antes de entrar
        acumulado += servicio
    return total / len(orden)

print(f"Espera media por orden de llegada     : {espera_acumulada(orden_llegada):>7.1f} min")
print(f"Espera media atendiendo al más corto  : {espera_acumulada(orden_corta_primero):>7.1f} min")
print()
peor_llegada = max(orden_llegada, key=lambda t: t[2])
print("Con la segunda política, el paciente de", peor_llegada[1],
      f"con {peor_llegada[2]:.1f} min de atención queda de último.")
````

El promedio mejora de forma notable. Y el paciente cuya atención es más larga, que con frecuencia es también el más grave, pasa a ser el último de la fila. **El indicador mejoró y alguien concreto está peor.**

Esa tensión no se resuelve con código. Es una decisión de política pública sobre la que un equipo de datos puede informar, y que no le corresponde tomar en solitario.

🔬 **Nota técnica.** La política de atender primero al trabajo más corto (*shortest job first*) minimiza demostrablemente el tiempo medio de espera y produce **inanición** (*starvation*) de los trabajos largos: si siguen llegando trabajos cortos, el largo no se atiende nunca. Los sistemas reales lo mitigan con envejecimiento, subiendo la prioridad de lo que lleva mucho esperando. En salud, la respuesta institucional se llama triaje, y ordena por gravedad clínica y no por duración, lo que empeora el promedio a propósito.

### 7.1 La comprensión de lista

Una comprensión describe **qué contiene** la lista nueva, en lugar de detallar los pasos para construirla.

````python
# Con bucle, como venías haciéndolo
en_horas_bucle = []
for espera in esperas:
    en_horas_bucle.append(round(espera / 60, 2))

# Con comprensión: la misma lista, en una línea
en_horas = [round(espera / 60, 2) for espera in esperas]

print("¿Producen el mismo resultado?", en_horas == en_horas_bucle)
print("Primeros tres:", en_horas[:3])
````

La estructura es `[expresión for elemento in secuencia]`, y se lee de dentro hacia fuera: para cada `espera` de `esperas`, calcula `round(espera / 60, 2)` y ponlo en la lista.

Con una condición, la comprensión también filtra.

````python
criticas = [espera for espera in esperas if espera > 60]
print("Citas con espera mayor a 60 min :", len(criticas))
print(f"Porcentaje del total            : {len(criticas) / len(esperas) * 100:.1f} %")

# Filtrar sobre tuplas: la condición puede mirar cualquier campo
ortopedia_criticas = [c for c in citas if c[1] == "Ortopedia" and c[0] > 60]
print("Citas críticas de Ortopedia     :", len(ortopedia_criticas))

# La expresión puede transformar a la vez que filtra
horas_pico = [c[2] for c in citas if c[0] > 90]
print("Horas de las citas peores       :", sorted(set(horas_pico)))
````

⚠️ **Error frecuente.** Confundir el lugar de la condición. Escrita al final, `[x for x in lista if cond]`, la condición **filtra**: los elementos que no la cumplen no aparecen. Escrita al principio como expresión condicional, `[x if cond else otro for x in lista]`, **transforma** y conserva todos los elementos. La primera produce una lista más corta; la segunda, una del mismo tamaño. Elegir mal no produce ningún error, solo un tamaño inesperado, que es la razón por la que este cuaderno insiste en verificarlo.

### 7.2 `map`, `filter` y `zip`

````python
# map aplica una función a cada elemento
en_horas_map = list(map(lambda e: round(e / 60, 2), esperas))
print("map    :", en_horas_map[:3])

# filter conserva los elementos que cumplen una condición
criticas_filter = list(filter(lambda e: e > 60, esperas))
print("filter : ", len(criticas_filter), "citas críticas")

# zip empareja elemento a elemento varias secuencias
primeras = list(zip(esperas[:3], especialidades[:3], horas[:3]))
print("zip    :", primeras)

# Y ahora la línea que la sección 4 construyó con un bucle entero:
citas_zip = list(zip(esperas, especialidades, horas, dias))
print("\n¿zip reproduce la lista de citas de la sección 4?", citas_zip == citas)
````

Esa última comprobación merece atención. Las seis líneas del bucle de la sección 4.4 se reducen a una llamada a `zip`, y el resultado es idéntico, tuplas incluidas. Se hizo primero con bucle a propósito: entender qué construye `zip` es más fácil después de haberlo construido a mano.

🔬 **Nota técnica.** `map`, `filter` y `zip` no devuelven listas sino **iteradores perezosos**: no calculan nada hasta que alguien recorre el resultado, y se agotan tras un solo recorrido. Eso permite encadenarlos sobre secuencias enormes sin materializar los intermedios, y explica por qué `print(zip(a, b))` muestra un objeto en lugar de datos, y por qué recorrer dos veces el mismo `map` devuelve vacío la segunda. La conversión con `list()` fuerza el cálculo. Otra consecuencia menos conocida: `zip` se detiene en la secuencia más corta y **descarta el resto sin avisar**, de modo que emparejar listas de longitudes distintas produce una pérdida silenciosa de datos; desde Python 3.10, `zip(a, b, strict=True)` lanza `ValueError` en ese caso, y conviene usarlo siempre que las longitudes deban coincidir.

### 7.3 Cuándo una comprensión deja de ayudar

````python
# Legible: una transformación, un filtro
legible = [c[0] for c in citas if c[1] == "Pediatría"]
print("Citas de Pediatría:", len(legible))

# Ilegible: tres condiciones, una transformación condicional y un anidamiento
ilegible = [round(c[0] / 60, 2) if c[0] > 60 else 0 for c in citas
            if c[1] in ("Ortopedia", "Pediatría") and c[2] >= 12 and c[3] != "Lunes"]
print("Resultado del monstruo:", len(ilegible), "elementos")

# La misma lógica, escrita para que otra persona la entienda
resultado = []
for espera, especialidad, hora, dia in citas:
    if especialidad not in ("Ortopedia", "Pediatría"):
        continue
    if hora < 12 or dia == "Lunes":
        continue
    resultado.append(round(espera / 60, 2) if espera > 60 else 0)

print("¿Coinciden?", resultado == ilegible)
````

Las dos versiones producen exactamente lo mismo. La segunda ocupa siete líneas en lugar de dos y **es mejor código**, porque cada condición está en su propia línea, se puede comentar por separado y un error de lógica se localiza sin releer la expresión completa.

La regla práctica que conviene adoptar: **una comprensión con más de una condición, o con una expresión condicional dentro, probablemente debía ser un bucle**. La elegancia que no se puede leer en voz alta no es elegancia.

### 8.1 De la posición a la clave

Todas las estructuras del módulo hasta aquí se consultan por **posición**: hay que saber que la cita que interesa está en el índice 417. Un diccionario se consulta por **clave**: un identificador que significa algo.

````python
# Un diccionario asocia claves con valores
recargo = {
    "Medicina interna": 0,
    "Pediatría": 6,
    "Ortopedia": 14,
    "Ginecología": 3,
}

print("Diccionario completo :", recargo)
print("Recargo de Ortopedia :", recargo["Ortopedia"], "min")
print("Número de claves     :", len(recargo))

recargo["Dermatología"] = 5          # añadir una clave nueva
recargo["Ortopedia"] = 16            # modificar una existente
print("\nTras añadir y modificar:", recargo)

del recargo["Dermatología"]
print("Tras eliminar          :", list(recargo.keys()))
````

🌱 **Analogía.** Una lista es una fila de casilleros numerados: para encontrar algo hay que recordar el número. Un diccionario es un archivador con pestañas rotuladas: se busca por el rótulo, y da igual en qué orden estén las carpetas. La consecuencia práctica es que un diccionario responde igual de rápido con diez claves que con diez millones, mientras que buscar un valor en una lista obliga a recorrerla.

🔬 **Nota técnica.** Un diccionario de Python está implementado como una tabla de dispersión (*hash table*): la clave se convierte en un número que indica dónde mirar, de modo que la búsqueda cuesta tiempo aproximadamente constante en lugar de proporcional al tamaño. Desde Python 3.7 el lenguaje **garantiza** que la iteración respeta el orden de inserción, algo que antes era un detalle de implementación y no una promesa. Esa garantía es cómoda y conviene no apoyarse en ella para razonar: un diccionario sigue siendo una estructura de acceso por clave, y si el orden importa, conviene declararlo de forma explícita al recorrerlo.

### 8.2 La clave que no existe

Acceder a una clave inexistente lanza `KeyError`, que es un error ruidoso y por tanto benigno.

````python
recargo = {"Medicina interna": 0, "Pediatría": 6, "Ortopedia": 14, "Ginecología": 3}
print(recargo["Neurología"])
````

````python
recargo = {"Medicina interna": 0, "Pediatría": 6, "Ortopedia": 14, "Ginecología": 3}

# get() devuelve None, o el valor por defecto que se le indique
print("get sin defecto :", recargo.get("Neurología"))
print("get con defecto :", recargo.get("Neurología", 0), "min")
print("Clave existente :", recargo.get("Ortopedia", 0), "min")

# in comprueba la pertenencia de la CLAVE, no del valor
print("\n¿Está Ortopedia? :", "Ortopedia" in recargo)
print("¿Está el valor 14?:", 14 in recargo, "· in mira las claves")
print("¿Está en valores? :", 14 in recargo.values())
````

⚠️ **Error frecuente.** Usar `get` con valor por defecto donde correspondía un `KeyError`. `recargo.get(especialidad, 0)` es cómodo y convierte una especialidad desconocida en un recargo de cero, en silencio. Si la especialidad desconocida es un error de captura, el análisis continúa con un dato inventado. La regla: `get` con defecto cuando la ausencia es **esperada y significativa**; acceso directo cuando la ausencia sería un **fallo** que conviene detener.

## 9. Conteo de frecuencias y anidamiento · Frequency counting and nesting

**Al terminar esta sección podrás:** contar categorías con diccionarios y con `Counter`, acumular sumas y medias por grupo, construir y recorrer diccionarios anidados, ordenar un diccionario por sus valores, y explicar por qué el promedio de los promedios no es el promedio.

### 9.1 Contar a mano

````python
# El patrón de conteo con diccionario: la operación más frecuente del análisis de datos
frecuencia = {}
for espera, especialidad, hora, dia in citas:
    frecuencia[especialidad] = frecuencia.get(especialidad, 0) + 1

print("Citas por especialidad:")
for especialidad, n in frecuencia.items():
    print(f"  {especialidad:<18} {n:>4}   {n / len(citas) * 100:>5.1f} %")
print(f"  {'TOTAL':<18} {sum(frecuencia.values()):>4}")
````

`frecuencia.get(clave, 0) + 1` es el modismo central: si la clave existe devuelve su valor, si no devuelve cero, y en ambos casos suma uno. Sin él haría falta un `if` para distinguir la primera aparición de las siguientes.

### 9.2 `Counter`: el conteo ya resuelto

````python
from collections import Counter

conteo = Counter(especialidades)

print("Counter completo :", dict(conteo))
print("Más frecuente    :", conteo.most_common(1))
print("Ranking completo :", conteo.most_common())
print("Clave inexistente:", conteo["Neurología"], "· devuelve 0 en lugar de KeyError")

# Counter también cuenta valores numéricos agrupados
por_hora = Counter(horas)
print("\nCitas por hora, de mayor a menor:")
for hora, n in por_hora.most_common(3):
    print(f"  {hora}:00 → {n} citas")
````

🔬 **Nota técnica.** `Counter` es una subclase de `dict` con tres diferencias útiles: devuelve cero para claves ausentes en lugar de lanzar `KeyError`, ofrece `most_common` para obtener el ranking ya ordenado, y admite aritmética entre contadores, de modo que `a - b` compara dos periodos y `a + b` los consolida. Esa tolerancia ante claves ausentes es cómoda y tiene el mismo riesgo que `get` con valor por defecto: una categoría mal escrita no produce error, produce un cero.

### 9.3 Acumular sumas y medias por grupo

Contar es el caso fácil. Promediar por grupo exige llevar dos acumuladores a la vez.

````python
suma_por_esp = {}
n_por_esp = {}

for espera, especialidad, hora, dia in citas:
    suma_por_esp[especialidad] = suma_por_esp.get(especialidad, 0) + espera
    n_por_esp[especialidad] = n_por_esp.get(especialidad, 0) + 1

media_por_esp = {}
for especialidad in suma_por_esp:
    media_por_esp[especialidad] = suma_por_esp[especialidad] / n_por_esp[especialidad]

print("Espera media por especialidad, de mayor a menor:")
for especialidad, media in sorted(media_por_esp.items(), key=lambda par: par[1], reverse=True):
    print(f"  {especialidad:<18} {media:>6.1f} min   ({n_por_esp[especialidad]} citas)")

media_global = sum(esperas) / len(esperas)
print(f"\n  {'GLOBAL':<18} {media_global:>6.1f} min   ({len(esperas)} citas)")
````

Ese es el segundo hallazgo del módulo, y ya es accionable: **Ortopedia espera sistemáticamente más que las demás especialidades**, y la diferencia no es de un par de minutos.

Fíjate también en la forma de ordenar un diccionario por sus valores: `sorted(dic.items(), key=lambda par: par[1])`. `items()` produce pares, y la clave de ordenamiento selecciona el segundo elemento de cada par.

⚠️ **Error frecuente y grave.** Calcular el promedio general como promedio de los promedios de grupo. Solo coincide cuando todos los grupos tienen exactamente el mismo tamaño; en cualquier otro caso da un valor distinto, porque da el mismo peso a un grupo de mil observaciones que a uno de tres. Compruébalo.

````python
media_de_medias = sum(media_por_esp.values()) / len(media_por_esp)
media_ponderada = sum(esperas) / len(esperas)

print(f"Promedio de los promedios : {media_de_medias:.3f} min")
print(f"Promedio real (ponderado) : {media_ponderada:.3f} min")
print(f"Diferencia                : {abs(media_de_medias - media_ponderada):.3f} min")
print()
print("Aquí la diferencia es pequeña porque los cuatro grupos tienen tamaños")
print("parecidos:", sorted(n_por_esp.values()))
print("Con grupos muy desiguales, el mismo error produce diferencias enormes.")
````

### 9.4 Diccionarios anidados: dos criterios a la vez

````python
# Estructura: especialidad → hora → número de citas
anidado = {}
for espera, especialidad, hora, dia in citas:
    if especialidad not in anidado:
        anidado[especialidad] = {}
    anidado[especialidad][hora] = anidado[especialidad].get(hora, 0) + 1

print("Ortopedia, citas por hora:")
for hora in sorted(anidado["Ortopedia"]):
    print(f"  {hora}:00 → {anidado['Ortopedia'][hora]}")

# Recorrido de los dos niveles
total = 0
for especialidad, por_hora in anidado.items():
    total += sum(por_hora.values())
print("\nTotal recorriendo los dos niveles:", total)
````

🔬 **Nota técnica.** El anidamiento resuelve el problema y se vuelve incómodo enseguida: cada nivel obliga a comprobar si la clave existe antes de escribir, y consultar exige encadenar accesos que fallan si falta cualquier eslabón. Hay dos alternativas mejores. La primera es `collections.defaultdict`, que crea el nivel ausente automáticamente. La segunda, más limpia para dos criterios, es la **clave compuesta** de la sección 8.4: un solo diccionario con claves `(especialidad, hora)`, plano y sin anidamiento. La regla práctica es que a partir de dos niveles conviene la clave compuesta, y a partir de tres el problema pide una estructura tabular, que es donde entra `DataFrame` en el Módulo 3.

### 9.5 🕵️ Error silencioso 6 · el diccionario que recuerda entre llamadas

````python
def contar_dias(lista_citas, acumulador={}):      # ← el valor por defecto es la trampa
    '''Cuenta citas por día y devuelve el acumulador.'''
    for espera, especialidad, hora, dia in lista_citas:
        acumulador[dia] = acumulador.get(dia, 0) + 1
    return acumulador


primeras_100 = contar_dias(citas[:100])
print("Primeras 100 citas :", sum(primeras_100.values()), "contadas")

siguientes_100 = contar_dias(citas[100:200])
print("Siguientes 100     :", sum(siguientes_100.values()), "contadas")
print("¿Deberían ser 100? Sí. ¿Lo son?")
````

<details><summary><b>Ver diagnóstico y solución</b></summary>

La segunda llamada informa 200 y no 100. El valor por defecto de un parámetro **se evalúa una sola vez, al definir la función**, de modo que las dos llamadas comparten el mismo diccionario: la segunda empieza con los conteos de la primera ya dentro.

Es el error más famoso de Python y sigue apareciendo en código de producción, porque el resultado es plausible y nada falla. La corrección es idiomática y conviene memorizarla:

```
def contar_dias(lista_citas, acumulador=None):
    if acumulador is None:
        acumulador = {}
    ...
```

Con `None` como valor por defecto, cada llamada que no reciba acumulador crea uno nuevo, y quien sí quiera acumular entre llamadas puede pasarlo de forma explícita, que es como debía haberse hecho desde el principio.

La regla general: **nunca uses un objeto mutable como valor por defecto de un parámetro**. Listas, diccionarios y conjuntos quedan excluidos; números, cadenas, tuplas y `None` son seguros.
</details>

## 10. Conjuntos y comparación de categorías · Sets

**Al terminar esta sección podrás:** crear conjuntos y usarlos para eliminar duplicados, aplicar unión, intersección y diferencia para comparar dos fuentes de datos, aprovechar la velocidad de la comprobación de pertenencia, y reconocer cuándo eliminar duplicados destruye información.

### 10.1 Un conjunto no tiene orden ni repetidos

````python
# Crear conjuntos
especialidades_unicas = set(especialidades)
horas_unicas = set(horas)

print("Especialidades distintas :", especialidades_unicas)
print("Cuántas                  :", len(especialidades_unicas))
print("Horas distintas          :", sorted(horas_unicas))
print()
print("Citas registradas        :", len(especialidades))
print("Valores distintos        :", len(especialidades_unicas))
print("Duplicados eliminados    :", len(especialidades) - len(especialidades_unicas))
````

Ese último número es la advertencia de toda la sección. Convertir a conjunto eliminó 896 elementos, y en este caso era exactamente lo que queríamos: nos preguntábamos **cuáles** especialidades existen, no cuántas citas hay. La misma operación aplicada a los tiempos de espera sería un desastre, y lo veremos en la sección 10.5.

### 10.2 Las operaciones de conjuntos

Aquí es donde los conjuntos dejan de ser una curiosidad y se vuelven la herramienta correcta para comparar dos fuentes de datos.

````python
# Simulamos el catálogo oficial y lo efectivamente registrado
catalogo_oficial = {"Medicina interna", "Pediatría", "Ortopedia", "Ginecología", "Dermatología"}
en_el_registro = set(especialidades)

print("Catálogo oficial :", sorted(catalogo_oficial))
print("En el registro   :", sorted(en_el_registro))
print()
print("Unión            :", sorted(catalogo_oficial | en_el_registro))
print("Intersección     :", sorted(catalogo_oficial & en_el_registro))
print("En el catálogo y NO en el registro :", sorted(catalogo_oficial - en_el_registro))
print("En el registro y NO en el catálogo :", sorted(en_el_registro - catalogo_oficial))
print("Diferencia simétrica (en uno u otro, no en ambos):",
      sorted(catalogo_oficial ^ en_el_registro))
````

Las dos diferencias responden preguntas distintas y ambas importan en la fase 2 de CRISP-DM. **Lo que está en el catálogo y no en el registro** puede ser un servicio que no operó, o uno que operó y cuyos datos no llegaron. **Lo que está en el registro y no en el catálogo** es siempre una alerta: una categoría no prevista, casi siempre un error de captura o un cambio que nadie comunicó.

🌱 **Analogía.** Es la lista de invitados frente a la lista de asistentes. La intersección son quienes vinieron y estaban invitados; una diferencia son los que faltaron; la otra, los que llegaron sin estar en la lista, que es la que conviene revisar primero.

### 10.3 Pertenencia: donde el conjunto es imbatible

````python
import time

# Buscar en una lista obliga a recorrerla; buscar en un conjunto, no.
valores_lista = list(range(200_000))
valores_set = set(valores_lista)

inicio = time.perf_counter()
for objetivo in (5, 100_000, 199_999):
    objetivo in valores_lista
costo_lista = time.perf_counter() - inicio

inicio = time.perf_counter()
for objetivo in (5, 100_000, 199_999):
    objetivo in valores_set
costo_set = time.perf_counter() - inicio

print(f"Tres búsquedas en lista    : {costo_lista * 1000:>8.3f} ms")
print(f"Tres búsquedas en conjunto : {costo_set * 1000:>8.3f} ms")
print(f"El conjunto fue unas {costo_lista / max(costo_set, 1e-9):,.0f} veces más rápido")
````

🔬 **Nota técnica.** La comprobación `x in lista` recorre los elementos uno a uno y su costo crece con el tamaño; `x in conjunto` calcula la dispersión de `x` y mira directamente en la posición correspondiente, con un costo aproximadamente constante. La consecuencia práctica aparece en cuanto la comprobación se hace dentro de un bucle: filtrar un millón de registros contra una lista de exclusión de diez mil elementos cuesta del orden de diez mil millones de comparaciones, y convertir esa lista en conjunto lo reduce a un millón. Es una línea de código y varios órdenes de magnitud.

### 10.5 🕵️ Error silencioso 7 · el conjunto que borra la evidencia

````python
# "Limpiamos" las esperas eliminando valores repetidos antes de calcular
esperas_limpias = set(esperas)

media_original = sum(esperas) / len(esperas)
media_limpia = sum(esperas_limpias) / len(esperas_limpias)

print(f"Observaciones originales : {len(esperas)}")
print(f"Tras 'limpiar'           : {len(esperas_limpias)}")
print()
print(f"Media original           : {media_original:.2f} min")
print(f"Media tras 'limpiar'     : {media_limpia:.2f} min")
print(f"Diferencia               : {media_limpia - media_original:+.2f} min")
````

<details><summary><b>Ver diagnóstico y solución</b></summary>

La media cambió, y cambió hacia arriba. Al convertir las esperas en conjunto, cada valor repetido quedó representado una sola vez, de modo que las esperas frecuentes, que son las cortas y medias, perdieron todo su peso, mientras que las esperas extremas, que casi nunca se repiten, conservaron el suyo íntegro. El resultado es una media inflada calculada sobre un conjunto de valores que ya no representa a ninguna población.

El error de fondo es conceptual y no técnico: **la repetición era el dato**. Que doscientos pacientes esperaran treinta y cinco minutos no es redundancia que convenga eliminar, es la frecuencia que da forma a la distribución.

La regla que conviene retener: un conjunto responde a la pregunta *¿qué valores distintos existen?* y nunca a la pregunta *¿cómo se distribuyen las observaciones?*. Para la primera es la herramienta ideal; para la segunda destruye exactamente la información que hacía falta.

Deduplicar es legítimo cuando se eliminan **registros repetidos por un error de carga**, identificados por una clave única, como en la sección 10.4. Nunca cuando se eliminan **valores repetidos de una medición**.
</details>

### 11.1 Del contenedor flexible al bloque homogéneo

Todas las estructuras vistas hasta aquí comparten un rasgo: guardan **referencias a objetos** y admiten mezclar tipos. Esa flexibilidad es la que permite que una lista contenga un número, un texto y otra lista a la vez, y también la que hace que cada operación aritmética tenga que preguntar, elemento por elemento, con qué tipo está tratando.

Un `ndarray` de NumPy renuncia a esa flexibilidad a cambio de velocidad.

````python
import numpy as np

esperas_np = np.array(esperas)

print("Tipo del objeto :", type(esperas_np).__name__)
print("Tipo de dato    :", esperas_np.dtype)      # todos los elementos comparten UN tipo
print("Forma           :", esperas_np.shape)      # una dimensión, 900 elementos
print("Dimensiones     :", esperas_np.ndim)
print("Bytes por dato  :", esperas_np.itemsize)
print("Bytes en total  :", esperas_np.nbytes)
print("\nPrimeros cinco  :", esperas_np[:5])
````

Tres atributos nuevos que conviene mirar siempre que algo no cuadre: `dtype` dice de qué tipo son **todos** los elementos, `shape` dice cuántos hay en cada dimensión, y `ndim` cuántas dimensiones tiene.

🌱 **Analogía.** Una lista es un cajón con compartimentos que pueden contener cualquier cosa, cada uno con una etiqueta que indica qué guarda. Un arreglo es una tira de casillas idénticas, todas del mismo tamaño y del mismo tipo, sin etiquetas porque no hacen falta. Buscar en el cajón exige leer etiquetas; recorrer la tira solo exige avanzar de casilla en casilla.

### 11.2 Qué se gana: memoria y velocidad

````python
import sys
import time

n = 1_000_000
lista_grande = list(range(n))
array_grande = np.arange(n)

mem_lista = sys.getsizeof(lista_grande) + sum(sys.getsizeof(x) for x in lista_grande[:1000]) / 1000 * n
mem_array = array_grande.nbytes

print(f"Memoria aproximada de la lista  : {mem_lista / 1e6:>8.1f} MB")
print(f"Memoria del arreglo             : {mem_array / 1e6:>8.1f} MB")
print(f"Razón                           : {mem_lista / mem_array:>8.1f} veces")

# Calentamiento: la primera operación de NumPy paga costos de inicialización
_ = (np.arange(10) * 2).sum()

inicio = time.perf_counter()
suma_lista = 0
for x in lista_grande:               # bucle explícito, en el intérprete de Python
    suma_lista += x * 2
t_lista = time.perf_counter() - inicio

inicio = time.perf_counter()
suma_array = int((array_grande * 2).sum())   # una sola operación vectorizada
t_array = time.perf_counter() - inicio

print(f"\nDuplicar y sumar con lista      : {t_lista:>8.4f} s")
print(f"Duplicar y sumar con arreglo    : {t_array:>8.4f} s")
print(f"El arreglo fue {t_lista / t_array:.0f} veces más rápido")
print(f"¿Mismo resultado? {suma_lista == suma_array}")
````

🔬 **Nota técnica.** La diferencia tiene dos causas independientes. La primera es la **contigüidad**: un `ndarray` almacena los valores uno tras otro en un bloque de memoria, mientras que una lista guarda punteros a objetos dispersos, de modo que recorrerla implica saltos que anulan la caché del procesador. La segunda es que las operaciones de NumPy se ejecutan en código compilado en C sobre el bloque entero, sin volver al intérprete de Python en cada elemento. Por eso la ganancia aparece en operaciones sobre arreglos completos y **desaparece si se recorre un `ndarray` con un bucle de Python**, que suele ser más lento que recorrer una lista. La regla práctica: si escribiste un `for` sobre un arreglo, casi siempre hay una operación vectorizada que lo reemplaza.

### 11.3 Qué se paga: la homogeneidad

````python
import numpy as np

# Al mezclar tipos, NumPy NO falla: convierte todo al tipo más general
mezcla = np.array([240, 42.5, True])
print("Mezcla numérica :", mezcla, "· dtype", mezcla.dtype)

con_texto = np.array([240, 42.5, "Ortopedia"])
print("Con texto       :", con_texto, "· dtype", con_texto.dtype)
print("¿Se puede sumar el arreglo con texto?")
try:
    con_texto.sum()
except TypeError as fallo:
    print("   No:", fallo)
````

⚠️ **Error frecuente y silencioso.** Ese segundo caso es el equivalente exacto de la coerción de R que vimos en el cuaderno electivo: **un solo valor de texto convierte el arreglo entero en texto**, sin error ni advertencia. Si una columna de mil números llega con una única celda que dice «sin dato», el arreglo resultante es de cadenas, las operaciones aritméticas dejan de funcionar y el diagnóstico no es evidente. La comprobación barata es mirar el `dtype` inmediatamente después de crear cualquier arreglo a partir de datos externos.

Quedó `14`. El arreglo tiene `dtype` entero, y al asignarle un decimal NumPy lo **trunca hacia cero** en silencio, sin redondear siquiera. Conviene notar que el problema no existe en una lista de Python, que habría guardado el 14.9 tal cual: es el precio de la homogeneidad.

### 12.2 La difusión: operar entre formas distintas

````python
import numpy as np

# Un escalar se difunde sobre todo el arreglo: eso ya lo usamos
print("Arreglo menos escalar :", (np.arange(5) - 2))

# Dos arreglos de la misma forma operan elemento a elemento
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
print("Elemento a elemento   :", a + b)

# Y formas compatibles se expanden sin copiar los datos
matriz = np.arange(12).reshape(3, 4)
por_columna = np.array([100, 200, 300, 400])

print("\nMatriz 3x4:\n", matriz)
print("\nSumando un vector de 4 a cada fila:\n", matriz + por_columna)
````

````python
import numpy as np
matriz = np.arange(12).reshape(3, 4)
matriz + np.array([1, 2, 3])          # formas incompatibles: 4 columnas frente a 3
````

El error es `ValueError: operands could not be broadcast together`. La regla de difusión compara las formas dimensión por dimensión, de derecha a izquierda, y las considera compatibles si son iguales o si una de ellas vale uno. Aquí 4 y 3 no cumplen ninguna de las dos condiciones.

Es un error ruidoso, y conviene celebrarlo: el caso peligroso sería el contrario, que dos formas incompatibles se combinaran en silencio produciendo un resultado del tamaño equivocado.

### 12.3 Indexación booleana: filtrar sin bucles

````python
import numpy as np

esperas_np = np.array(esperas, dtype=float)
especialidades_np = np.array(especialidades)
horas_np = np.array(horas)

mascara = esperas_np > 60                # un arreglo de True y False, uno por cita
print("Tipo de la máscara :", mascara.dtype)
print("Primeras diez       :", mascara[:10])
print("Cuántas cumplen     :", mascara.sum(), "· sum() cuenta los True")
print("Proporción          :", f"{mascara.mean() * 100:.1f} %")

criticas = esperas_np[mascara]           # el arreglo filtrado
print("\nEsperas críticas    :", criticas.size, "valores")
print("Media de las críticas:", f"{criticas.mean():.1f} min")

# Condiciones combinadas: & y | en lugar de and y or, y paréntesis obligatorios
ortopedia_tarde = (especialidades_np == "Ortopedia") & (horas_np >= 13)
print("\nCitas de Ortopedia por la tarde :", ortopedia_tarde.sum())
print("Su espera media                  :", f"{esperas_np[ortopedia_tarde].mean():.1f} min")
````

⚠️ **Error frecuente.** Escribir `and` y `or` entre máscaras. Esos operadores esperan un único valor de verdad y con un arreglo lanzan `ValueError: The truth value of an array with more than one element is ambiguous`. Con arreglos hay que usar `&`, `|` y `~`, **y encerrar cada condición entre paréntesis**, porque `&` tiene mayor precedencia que los operadores de comparación y sin paréntesis la expresión se agrupa de forma inesperada.

### 12.4 `np.where`: transformar según una condición

````python
import numpy as np

esperas_np = np.array(esperas, dtype=float)

# Clasificar sin bucles ni comprensiones
categoria = np.where(esperas_np > 60, "critica",
                     np.where(esperas_np > 30, "alta", "aceptable"))

valores, conteos = np.unique(categoria, return_counts=True)
print("Distribución de categorías:")
for v, c in zip(valores, conteos):
    print(f"  {v:<10} {c:>4}   {c / esperas_np.size * 100:>5.1f} %")

# where también sirve para localizar posiciones
posiciones = np.where(esperas_np > 150)[0]
print("\nCitas con más de 150 min de espera:", posiciones.size)
print("Sus índices:", posiciones[:10])
````

### 12.5 🕵️ Error silencioso 9 · la segmentación que no copia

Esta es la diferencia entre listas y arreglos que más errores produce al pasar de unas a otros.

````python
import numpy as np

# Con LISTAS, la segmentación copia
lista_original = [10, 20, 30, 40, 50]
trozo_lista = lista_original[1:4]
trozo_lista[0] = 999
print("Lista original tras modificar el trozo :", lista_original)

# Con ARREGLOS, la segmentación es una VISTA sobre la misma memoria
array_original = np.array([10, 20, 30, 40, 50])
trozo_array = array_original[1:4]
trozo_array[0] = 999
print("Arreglo original tras modificar el trozo:", array_original)
print("\n¿El trozo comparte memoria con el original?",
      np.shares_memory(array_original, trozo_array))
````

<details><summary><b>Ver diagnóstico y solución</b></summary>

La lista quedó intacta y el arreglo cambió. `array_original[1:4]` no copia nada: devuelve una **vista**, una ventana sobre el mismo bloque de memoria, de modo que escribir en la vista escribe en el original.

La decisión de diseño es deliberada y correcta: copiar un tramo de un arreglo de diez millones de elementos cada vez que alguien lo segmenta sería ruinoso. Pero produce un error silencioso muy difícil de rastrear, porque quien modifica el trozo suele estar en otra función, a veces en otro archivo, y no tiene forma de saber que está tocando el original.

La solución es explícita: `trozo = array_original[1:4].copy()`. Y la comprobación, cuando haya dudas, es `np.shares_memory(a, b)`.

Conviene notar que la indexación booleana de la sección 12.3 **sí copia**: `esperas_np[mascara]` devuelve un arreglo nuevo, porque los elementos seleccionados no son contiguos y no hay ventana posible sobre ellos. Esa asimetría entre segmentación y máscara es la que hace que el comportamiento parezca arbitrario cuando no lo es.
</details>

### 13.2 La matriz del caso: especialidad por hora

````python
import numpy as np

esperas_np = np.array(esperas, dtype=float)
especialidades_np = np.array(especialidades)
horas_np = np.array(horas)

ESPEC = sorted(set(especialidades))
HORAS = sorted(set(horas))

matriz_conteo = np.zeros((len(ESPEC), len(HORAS)), dtype=int)
matriz_media = np.zeros((len(ESPEC), len(HORAS)))

for f, especialidad in enumerate(ESPEC):
    for c, hora in enumerate(HORAS):
        grupo = esperas_np[(especialidades_np == especialidad) & (horas_np == hora)]
        matriz_conteo[f, c] = grupo.size
        matriz_media[f, c] = grupo.mean() if grupo.size > 0 else np.nan

print("Espera media (min) por especialidad y hora")
print(" " * 18 + "".join(f"{h:>7}" for h in HORAS))
for f, especialidad in enumerate(ESPEC):
    print(f"{especialidad:<18}" + "".join(f"{v:>7.1f}" for v in matriz_media[f]))
print("\nTotal de citas en la matriz:", matriz_conteo.sum(), "de", esperas_np.size)
````

### 13.3 Agregar por ejes

Aquí está la idea central de la sección, y la que más confusión produce.

````python
import numpy as np

print("Forma de la matriz de conteos:", matriz_conteo.shape, "· (especialidades, horas)")
print()
print("matriz_conteo.sum()        →", matriz_conteo.sum(), "· todo, sin ejes")
print("matriz_conteo.sum(axis=0)  →", matriz_conteo.sum(axis=0))
print("   Colapsa las FILAS: queda un valor por hora")
print()
print("matriz_conteo.sum(axis=1)  →", matriz_conteo.sum(axis=1))
print("   Colapsa las COLUMNAS: queda un valor por especialidad")
print()
for f, especialidad in enumerate(ESPEC):
    print(f"  {especialidad:<18} {matriz_conteo.sum(axis=1)[f]:>4} citas")
````

🌱 **Analogía.** `axis` no nombra el eje que se conserva sino **el que desaparece**. Con `axis=0` desaparece la dimensión de las filas: todas las especialidades se funden y queda un total por hora. Con `axis=1` desaparece la de las columnas: todas las horas se funden y queda un total por especialidad. Si dudas, mira la forma del resultado: sumar sobre `axis=0` en una matriz de cuatro por diez deja diez números, no cuatro.

🔬 **Nota técnica.** La regla se generaliza a cualquier número de dimensiones: agregar sobre un eje elimina esa posición de la tupla `shape`. En un arreglo de forma `(4, 10, 3)`, sumar con `axis=1` produce forma `(4, 3)`. El parámetro `keepdims=True` conserva la dimensión con tamaño uno, lo que resulta útil precisamente para difundir el resultado de vuelta contra el arreglo original, como al restar a cada fila su propia matriz_media.

### 13.4 Localizar el máximo de una matriz

````python
import numpy as np

plano = np.nanargmax(matriz_media)                      # índice en la versión aplanada
f, c = np.unravel_index(plano, matriz_media.shape)      # lo convierte a fila y columna

print(f"Peor combinación : {ESPEC[f]} a las {HORAS[c]}:00")
print(f"Espera media     : {matriz_media[f, c]:.1f} min sobre {matriz_conteo[f, c]} citas")

print("\nPeor hora de cada especialidad:")
for i, especialidad in enumerate(ESPEC):
    j = int(np.nanargmax(matriz_media[i]))
    print(f"  {especialidad:<18} {HORAS[j]}:00 → {matriz_media[i, j]:>5.1f} min "
          f"({matriz_conteo[i, j]} citas)")
````

### 14.1 La figura y los ejes

Matplotlib ofrece dos interfaces. La rápida, `plt.plot()`, dibuja sobre una figura implícita y es cómoda para explorar. La de objetos, que crea explícitamente la figura y sus ejes, es la que conviene usar en cuanto la figura tenga más de un panel o vaya a un informe.

````python
import matplotlib.pyplot as plt
import numpy as np

esperas_np = np.array(esperas, dtype=float)

fig, ax = plt.subplots(figsize=(9, 4))          # fig es el lienzo, ax el panel
ax.hist(esperas_np, bins=40, color="#750ea5", edgecolor="white")
ax.set_title("Distribución del tiempo de espera · consulta externa, trimestre")
ax.set_xlabel("Minutos de espera")
ax.set_ylabel("Número de citas")
plt.tight_layout()
plt.show()
````

Ahí está, por fin, **el hallazgo central del módulo hecho visible**. La distribución no es simétrica: se concentra a la izquierda y arrastra una cola larga hacia la derecha. Esa cola es la que separa la media de la mediana, y es la razón por la que el promedio de cuarenta minutos que la gerencia viene discutiendo no describe la experiencia de la mayoría.

Un histograma responde a una pregunta que ningún indicador resumido puede responder: **cómo se reparten las observaciones**.

🔬 **Nota técnica.** El número de intervalos, `bins`, no es un detalle estético: cambia lo que la figura muestra. Con demasiados pocos, se ocultan modas y estructuras; con demasiados, el ruido de muestreo parece señal. Existen reglas automáticas, como la de Freedman y Diaconis, que ajustan la anchura según el rango intercuartílico y el tamaño de la muestra, y Matplotlib las acepta con `bins="fd"`. La práctica profesional consiste en probar dos o tres valores y comprobar que la conclusión no depende de esa elección; si depende, la conclusión no era sólida.

### 14.2 Marcar los estadísticos sobre la distribución

La distancia entre las dos primeras líneas **es** la asimetría. Y la tercera dice dónde está el problema: una de cada diez personas espera más que ese valor.

### 14.3 Barras: comparar categorías

````python
import matplotlib.pyplot as plt
import numpy as np

especialidades_np = np.array(especialidades)
ESPEC = sorted(set(especialidades))
medias_esp = [esperas_np[especialidades_np == e].mean() for e in ESPEC]
n_esp = [int((especialidades_np == e).sum()) for e in ESPEC]

orden = np.argsort(medias_esp)[::-1]
etiquetas = [f"{ESPEC[i]}\n(n={n_esp[i]})" for i in orden]
valores = [medias_esp[i] for i in orden]

fig, ax = plt.subplots(figsize=(9, 4.5))
barras = ax.bar(etiquetas, valores, color="#750ea5")
ax.axhline(esperas_np.mean(), color="#07a3b2", linestyle="--",
           label=f"Media global: {esperas_np.mean():.1f} min")
ax.bar_label(barras, fmt="%.1f", padding=3)
ax.set_ylim(0, max(valores) * 1.2)
ax.set_title("Espera media por especialidad")
ax.set_ylabel("Minutos")
ax.legend()
plt.tight_layout()
plt.show()
````

Dos decisiones de esa figura merecen atención porque no son cosméticas. La primera es que **el eje vertical empieza en cero**, lo que permite comparar las alturas de las barras entre sí. ### 14.5 🕵️ Error silencioso 11 · el eje que exagera

Las dos figuras siguientes muestran **exactamente los mismos datos**. Ejecuta la celda y compáralas.

````python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

ax1.bar(ESPEC, medias_esp, color="#750ea5")
ax1.set_ylim(0, max(medias_esp) * 1.15)
ax1.set_title("Eje desde cero")
ax1.set_ylabel("Minutos")
ax1.tick_params(axis="x", labelrotation=30, labelsize=8)

ax2.bar(ESPEC, medias_esp, color="#c1121f")
ax2.set_ylim(min(medias_esp) - 2, max(medias_esp) + 2)      # eje truncado
ax2.set_title("Eje truncado")
ax2.set_ylabel("Minutos")
ax2.tick_params(axis="x", labelrotation=30, labelsize=8)

plt.tight_layout()
plt.show()

print("Diferencia real entre la peor y la mejor especialidad:",
      f"{max(medias_esp) - min(medias_esp):.1f} min")
````

El truncamiento del eje vertical en un gráfico de barras rompe la correspondencia entre la longitud de la barra y el valor que representa, que es justamente lo que el lector interpreta sin pensarlo. La barra deja de ser una magnitud y pasa a ser una diferencia respecto de un origen arbitrario que nadie eligió con criterio.

La regla profesional es clara: **en un gráfico de barras, el eje empieza en cero, sin excepciones**. Si las diferencias son demasiado pequeñas para verse desde cero, la figura correcta no es una barra truncada sino un gráfico de puntos, que no invita a leer longitudes, o directamente la tabla de valores.

Este error no siempre es malintencionado. Muchas bibliotecas ajustan el eje automáticamente al rango de los datos, de modo que la exageración aparece **por defecto** y quien construye la figura ni siquiera la decide. Revisar el origen del eje antes de publicar es parte del oficio.
</details>

## Errores frecuentes del módulo

Los diez errores silenciosos que recorriste, en una tabla para volver a ella cuando algo no cuadre.

| # | Síntoma | Causa |
|---|---|---|
| 1 | El original cambió aunque solo tocaste la copia | `b = a` crea un alias, no una copia |

| 8 | La media sube al «limpiar» duplicados | La repetición era el dato |
| 9 | Un decimal guardado aparece truncado | El `dtype` del arreglo es entero |
| 10 | El resultado tiene diez valores y la pregunta pedía cuatro | Se agregó sobre el eje equivocado |
