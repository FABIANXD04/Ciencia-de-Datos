````python
# ---------------------------------------------------------------------
# Generación del registro trimestral SUCIO de consulta externa
# Datos SINTÉTICOS con semilla fija, pero con los defectos de un dato real:
# especialidades inconsistentes, fechas como texto, faltantes, tipos mezclados
# y filas duplicadas. Tu trabajo en este módulo es limpiarlos.
# ---------------------------------------------------------------------
import random

rng = random.Random(SEMILLA)

# La misma especialidad escrita de varias formas, como la teclean personas distintas
VARIANTES = {
    "Medicina interna": ["Medicina interna", "MEDICINA INTERNA", " medicina interna ", "Med. Interna"],
    "Pediatría":        ["Pediatría", "PEDIATRIA", "pediatria ", "Pediatria"],
    "Ortopedia":        ["Ortopedia", "ORTOPEDIA", " ortopedia", "ortopedia"],
    "Ginecología":      ["Ginecología", "GINECOLOGIA", "ginecologia ", "Ginecología "],
}
DIAS = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]
MESES = {"Lunes": "01", "Martes": "01", "Miércoles": "01", "Jueves": "01", "Viernes": "01"}
N = 900

filas = []
for i in range(N):
    hora = rng.randrange(7, 17)
    esp_limpia = rng.choice(list(VARIANTES))
    especialidad = rng.choice(VARIANTES[esp_limpia])      # se guarda una variante "sucia"
    dia = rng.choice(DIAS)

    base = 12 + (hora - 7) * 3.1
    base += {"Medicina interna": 0, "Pediatría": 6, "Ortopedia": 14, "Ginecología": 3}[esp_limpia]
    if dia == "Lunes":
        base += 9
    espera = base + rng.gauss(0, 7)
    if rng.random() < 0.08:
        espera += rng.uniform(40, 120)
    espera = round(max(espera, 1.0), 1)

    # La fecha llega como TEXTO, en formato día/mes/año
    dia_mes = rng.randint(1, 28)
    fecha_txt = f"{dia_mes:02d}/{MESES[dia]}/2026"

    # El tiempo de espera llega como TEXTO, con coma decimal a veces
    espera_txt = str(espera).replace(".", ",") if rng.random() < 0.5 else str(espera)

    # Un 6 % de las esperas viene VACÍA (no se capturó)
    if rng.random() < 0.06:
        espera_txt = ""

    filas.append([f"C{i:04d}", especialidad, str(hora), dia, fecha_txt, espera_txt])

# Añadimos 40 filas DUPLICADAS por un error de carga
for _ in range(40):
    filas.append(list(rng.choice(filas[:N])))

rng.shuffle(filas)

COLUMNAS = ["id_cita", "especialidad", "hora", "dia", "fecha", "espera_min"]
print("Filas generadas (incluye duplicados):", len(filas))
print("Columnas:", COLUMNAS)
print("\nCinco filas de muestra, tal como llegan:")
for fila in filas[:5]:
    print("  ", fila)
````

### 1.2 Los métodos esenciales de limpieza

````python
sucio = "  MEDICINA interna  "

print("Original          :", f"[{sucio}]")
print(".strip()          :", f"[{sucio.strip()}]")           # quita espacios de los extremos
print(".lower()          :", f"[{sucio.lower()}]")           # todo a minúsculas
print(".upper()          :", f"[{sucio.upper()}]")           # todo a mayúsculas
print(".strip().title()  :", f"[{sucio.strip().title()}]")   # Primera Letra De Cada Palabra
print(".replace()        :", f"[{sucio.replace(' ', '_')}]") # sustituye un texto por otro

# Los métodos se encadenan: cada uno devuelve una cadena nueva sobre la que sigue el siguiente
limpio = sucio.strip().lower().capitalize()
print("\nEncadenado        :", f"[{limpio}]")
````

Cada método **devuelve una cadena nueva** y deja intacta la original, porque las cadenas son **inmutables**, igual que las tuplas del Módulo 2. Por eso se pueden encadenar: `sucio.strip().lower()` primero recorta y sobre el resultado pone en minúsculas.

🔬 **Nota técnica.** La inmutabilidad de las cadenas tiene una consecuencia de rendimiento que aparece al construir texto dentro de un bucle: `resultado += pieza` crea una cadena nueva en cada vuelta, con costo cuadrático. La forma eficiente es acumular las piezas en una lista y unirlas al final con `"".join(lista)`, que recorre una sola vez. Con textos cortos da igual; al procesar millones de líneas, la diferencia es de minutos frente a horas.

### 1.4 Normalizar: de dieciséis categorías a cuatro

````python
def normalizar_especialidad(texto):
    '''Reduce una variante sucia a su forma canónica.

    Aplica strip y lower para unificar, y luego mapea a la forma correcta.
    '''
    t = texto.strip().lower()
    t = t.replace("á", "a").replace("í", "i")       # quita las tildes que estorban
    if t.startswith("med"):
        return "Medicina interna"
    if t.startswith("pedia"):
        return "Pediatría"
    if t.startswith("ortop"):
        return "Ortopedia"
    if t.startswith("gineco"):
        return "Ginecología"
    return "DESCONOCIDA"                             # marca lo que no encaja, no lo oculta


normalizadas = [normalizar_especialidad(e) for e in especialidades_crudas]
distintas_limpias = set(normalizadas)

print("Antes  :", len(set(especialidades_crudas)), "categorías")
print("Después :", len(distintas_limpias), "categorías:", sorted(distintas_limpias))
print("¿Quedó alguna sin reconocer?:", "DESCONOCIDA" in distintas_limpias)
````

Fíjate en la última rama de la función: lo que no encaja en ninguna categoría conocida se marca como `"DESCONOCIDA"` en lugar de descartarse o forzarse a una categoría cualquiera. **Un dato que no entiendes no se esconde: se señala.** Si más adelante aparece una especialidad nueva, el análisis lo dirá en voz alta en vez de asignarla en silencio a la categoría equivocada.

### 2.1 f-strings: el formato que ya conoces, a fondo

Las f-strings aparecieron en el Módulo 1. Aquí las llevamos hasta donde hacen falta para construir informes.

````python
especialidad = "Ortopedia"
media = 49.83521
n = 244
pct = 0.2711

print(f"Sin formato : {especialidad} {media} {pct}")
print(f"Un decimal  : {media:.1f} min")
print(f"Porcentaje  : {pct:.1%}")                 # :.1% multiplica por 100 y añade el signo
print(f"Ancho fijo  : [{especialidad:<20}] [{n:>6}]")   # < izquierda, > derecha
print(f"Miles       : {1234567:,}")               # separador de miles
print(f"Relleno     : {n:0>5}")                   # rellena con ceros a la izquierda
````

Los tres controles que usarás casi siempre: `:.1f` fija decimales, `:>8` alinea en un ancho dado, `:,` separa los miles. Y `:.1%` es el atajo para porcentajes: toma una proporción, la multiplica por cien y le pone el signo.

### 2.2 Construir un informe de texto

````python
def linea_reporte(especialidad, media, n, meta=30):
    '''Devuelve una línea de informe alineada para una especialidad.'''
    exceso = media - meta
    estado = "excede" if exceso > 0 else "cumple"
    return f"{especialidad:<18} {n:>4} citas  media {media:>5.1f} min  ({estado} por {abs(exceso):.1f})"


datos = [("Ortopedia", 49.8, 244), ("Ginecología", 38.9, 218),
         ("Pediatría", 38.6, 225), ("Medicina interna", 32.4, 213)]

print("INFORME DE ESPERA POR ESPECIALIDAD")
print("=" * 60)
for especialidad, media, n in datos:
    print(linea_reporte(especialidad, media, n))
````

🔬 **Nota técnica.** Desde Python 3.8, la f-string admite el sufijo `=` para depuración: `f"{media=}"` imprime a la vez el nombre de la variable y su valor, `media=49.8`. Es un atajo poco conocido y muy útil para inspeccionar el estado sin escribir el nombre dos veces. Y desde Python 3.12 las f-strings admiten comillas anidadas del mismo tipo y expresiones de varias líneas, lo que elimina restricciones que antes obligaban a rodeos.

### 3.1 Qué es una expresión regular

Una expresión regular es un patrón que describe una familia de textos. En lugar de buscar una cadena exacta, buscas "cualquier texto que tenga esta forma": una fecha, un correo, un código de cita. El módulo `re` de Python las implementa.

````python
import re

# El registro tiene identificadores como C0042 y fechas como 15/01/2026
texto = "La cita C0042 se agendó el 15/01/2026 a las 14 horas"

# \d significa "un dígito"; \d{4} significa "cuatro dígitos seguidos"
id_cita = re.search(r"C\d{4}", texto)
fecha = re.search(r"\d{2}/\d{2}/\d{4}", texto)

print("Identificador encontrado :", id_cita.group())
print("Fecha encontrada         :", fecha.group())
print("Posición del id          :", id_cita.span())
````

Los ladrillos básicos: `\d` es un dígito, `\d{4}` son cuatro dígitos seguidos, `\w` es una letra o dígito, `.` es cualquier carácter, `+` significa "uno o más" y `*` "cero o más". Con eso se describe la mayoría de los patrones que aparecen en datos.

⚠️ **La `r` antes de la comilla no es opcional.** `r"C\d{4}"` es una cadena cruda (*raw string*): le dice a Python que no interprete la barra invertida. Sin la `r`, `\d` podría confundirse con una secuencia de escape. En expresiones regulares, usa siempre `r"..."`.

### 3.2 Extraer, validar y reemplazar

````python
import re

fechas = ["15/01/2026", "3/1/2026", "2026-01-15", "sin fecha", "31/12/2026"]

patron_fecha = r"^\d{2}/\d{2}/\d{4}$"      # ^ inicio, $ fin: la cadena ENTERA debe cumplir

print("Validación de formato dd/mm/aaaa:")
for f in fechas:
    valida = re.match(patron_fecha, f) is not None
    print(f"  [{f:<12}] {'válida' if valida else 'NO cumple el formato'}")

# Extraer los tres componentes de una fecha con grupos
m = re.match(r"(\d{2})/(\d{2})/(\d{4})", "15/01/2026")
print("\nDía, mes, año:", m.group(1), m.group(2), m.group(3))

# Reemplazar: unificar separadores
print("Unificado:", re.sub(r"[/.]", "-", "15/01/2026"))   # / o . se vuelven -
````

Los paréntesis crean **grupos** que capturan partes del patrón, y `^` y `$` anclan el patrón al inicio y al fin de la cadena, de modo que la cadena entera deba cumplir, no solo una parte. Sin esos anclajes, `"basura 15/01/2026 basura"` pasaría la validación.

🔬 **Nota técnica.** Las expresiones regulares son poderosas y fáciles de convertir en un problema. Un patrón demasiado permisivo valida cosas que no debería; uno demasiado complejo se vuelve imposible de mantener. Existe una clase de patrones patológicos que provocan un tiempo de ejecución exponencial sobre ciertas entradas, un fallo conocido como *ReDoS*, que puede colgar un sistema. La regla práctica: para tareas simples de limpieza, prefiere los métodos de cadena de la sección 1; reserva las expresiones regulares para cuando el patrón sea genuinamente variable, y pruébalas con casos límite.

1. Anclan el patrón al inicio y al fin de la cadena, de modo que la cadena entera deba cumplirlo y no solo una parte. Validar formatos casi siempre exige ambos.
2. Para que Python no interprete la barra invertida como secuencia de escape; la cadena cruda pasa el patrón tal cual al módulo `re`.
3. Para tareas simples y fijas, como quitar espacios o pasar a minúsculas, el método de cadena es más claro y más rápido. La expresión regular se reserva para patrones genuinamente variables.
</details>

### 4.1 El orden de la limpieza importa

Limpiar no es aplicar operaciones al azar: el orden cambia el resultado. La celda siguiente muestra un flujo de limpieza sobre la columna de especialidades, paso a paso.

````python
def limpiar_especialidad(texto):
    '''Flujo completo de limpieza de una especialidad, paso a paso.'''
    paso1 = texto.strip()                    # 1. quitar espacios de los extremos
    paso2 = paso1.lower()                     # 2. unificar a minúsculas
    paso3 = paso2.replace("á", "a").replace("í", "i")   # 3. quitar tildes
    # 4. mapear a la forma canónica por prefijo
    if paso3.startswith("med"):
        return "Medicina interna"
    if paso3.startswith("pedia"):
        return "Pediatría"
    if paso3.startswith("ortop"):
        return "Ortopedia"
    if paso3.startswith("gineco"):
        return "Ginecología"
    return "DESCONOCIDA"


# Aplicamos y verificamos
crudas = [fila[1] for fila in filas]
limpias = [limpiar_especialidad(e) for e in crudas]

print("Categorías antes  :", len(set(crudas)))
print("Categorías después:", len(set(limpias)), "→", sorted(set(limpias)))
````

El orden no es arbitrario. Si intentaras mapear por prefijo **antes** de quitar espacios, `" ortopedia"` no empezaría por "ortop" sino por un espacio, y caería en "DESCONOCIDA". Cada paso prepara al siguiente: primero se quita lo que estorba, luego se unifica, y solo al final se clasifica.

🧩 **Trazado de estado.** Sigue una variante sucia por los cuatro pasos:

| Paso | Operación | Resultado sobre `" ORTOPEDIA "` |
|---|---|---|
| Inicio | (entrada cruda) | `" ORTOPEDIA "` |
| 1 | `.strip()` | `"ORTOPEDIA"` |
| 2 | `.lower()` | `"ortopedia"` |
| 3 | quitar tildes | `"ortopedia"` |
| 4 | mapear por prefijo | `"Ortopedia"` |

### 4.2 Un informe de calidad antes de analizar

La fase 2 de CRISP-DM produce un informe de calidad. Antes de tocar nada, se documenta qué tan sucios están los datos.

````python
def informe_calidad(filas, columnas):
    '''Produce un informe de calidad básico de un registro en filas.'''
    n = len(filas)
    print(f"REGISTRO: {n} filas, {len(columnas)} columnas")
    print("=" * 58)
    for j, col in enumerate(columnas):
        valores = [fila[j] for fila in filas]
        vacios = sum(1 for v in valores if v == "" or v is None)
        distintos = len(set(valores))
        print(f"  {col:<14} · {distintos:>4} distintos · {vacios:>3} vacíos "
              f"({vacios / n * 100:.1f} %)")


informe_calidad(filas, COLUMNAS)
````

Ese informe es el primer entregable de cualquier proyecto serio. Antes de calcular una sola media, dice cuántos valores faltan y cuántas categorías hay en cada columna. La columna de especialidad con dieciséis valores distintos grita "límpiame antes de agrupar"; la de espera con un 6 % de vacíos avisa de un problema que la sección 9 abordará con Pandas.

⚠️ **Error frecuente.** Analizar antes de conocer la calidad. Calcular la media de la columna de espera ahora mismo fallaría, porque los valores son texto con comas y hay celdas vacías. ¿Por qué el flujo de limpieza quita los espacios antes de mapear por prefijo, y no al revés?

<details><summary><b>Ver solución</b></summary>

Porque el mapeo por prefijo comprueba con qué empieza la cadena, y una cadena que empieza por un espacio, como `" ortopedia"`, no empieza por "ortop", así que caería en "DESCONOCIDA". El principio general es que la limpieza va de lo más superficial a lo más semántico: primero se elimina el ruido que estorba a cualquier operación posterior, luego se unifica el formato, y solo al final se interpreta el contenido.
</details>

````python
import pandas as pd

# read_csv tiene decenas de parámetros; estos resuelven la mayoría de los problemas reales.
# Los mostramos como referencia comentada, sobre un CSV de ejemplo con problemas típicos.
ejemplo = "nombre;edad;ciudad\nAna;30;Bogotá\nLuis;;Medellín\n"
with open("ejemplo.csv", "w", encoding="utf-8") as f:
    f.write(ejemplo)

# Sin parámetros, Pandas no entiende el separador de punto y coma
mal = pd.read_csv("ejemplo.csv")
print("Mal leído (una sola columna):", list(mal.columns))

# Con el separador correcto
bien = pd.read_csv("ejemplo.csv", sep=";")
print("Bien leído (tres columnas)  :", list(bien.columns))
print(bien)
````

### 7.2 loc e iloc: seleccionar filas

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)

# iloc: por POSICIÓN, como los índices de NumPy
print("Fila en posición 0 (iloc):")
print(df.iloc[0])

# loc: por ETIQUETA del índice y nombre de columna
print("\nEspecialidad de la fila con índice 5 (loc):")
print(df.loc[5, "especialidad"])

# loc admite rangos y listas de columnas
print("\nFilas 0 a 2, columnas id y especialidad (loc):")
print(df.loc[0:2, ["id_cita", "especialidad"]])
````

La distinción es esencial: **`iloc` selecciona por posición** (el número de fila, empezando en cero, como NumPy) y **`loc` selecciona por etiqueta** (el valor del índice y el nombre de la columna). Con un índice numérico por defecto coinciden a menudo, pero en cuanto el índice sea una fecha o un identificador, `loc` usa esa etiqueta y `iloc` sigue usando la posición.

🔬 **Nota técnica.** Un detalle que sorprende: `df.loc[0:2]` incluye la fila 2, mientras que `df.iloc[0:2]` la excluye. `loc` trabaja con etiquetas y su rango es inclusivo en ambos extremos; `iloc` trabaja con posiciones y sigue la convención de Python de excluir el fin. Mezclar los dos comportamientos es una fuente frecuente de errores de uno en uno, y por eso conviene tener siempre presente cuál de los dos se está usando.

### 7.3 Filtrar con máscaras booleanas

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)

# La máscara booleana es idéntica a la de NumPy del Módulo 2
mascara = df["dia"] == "Lunes"
print("Citas del lunes:", mascara.sum())

# Filtrar: el DataFrame se indexa con la máscara
lunes = df[df["dia"] == "Lunes"]
print("Forma del subconjunto del lunes:", lunes.shape)

# Combinar condiciones: & y |, cada una entre paréntesis (igual que NumPy)
lunes_ortopedia = df[(df["dia"] == "Lunes") & (df["especialidad"].str.contains("rtop"))]
print("Citas de ortopedia el lunes:", lunes_ortopedia.shape[0])
````

El filtrado de Pandas **es** la indexación booleana de NumPy que ya conoces, con una comodidad añadida: el accesor `.str` aplica métodos de cadena a toda una columna de texto de una vez. `df["especialidad"].str.contains("rtop")` busca ese fragmento en cada fila, sorteando de paso el problema de las mayúsculas al buscar solo la parte central del nombre.

Y las reglas son las mismas del Módulo 2: `&` y `|` en lugar de `and` y `or`, y cada condición entre paréntesis.

### 7.4 🕵️ Error silencioso 2 · la asignación encadenada

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)

# Queremos poner en 0 la hora de las citas de la primera fila del lunes.
# Esta forma encadenada es la trampa:
subconjunto = df[df["dia"] == "Lunes"]
subconjunto["hora"] = "MODIFICADO"        # ¿modifica df, o solo la copia?

print("¿El DataFrame original cambió?")
print("  Valores 'MODIFICADO' en df:", (df["hora"] == "MODIFICADO").sum())
print("  Valores 'MODIFICADO' en el subconjunto:", (subconjunto["hora"] == "MODIFICADO").sum())
````

<details><summary><b>Ver diagnóstico y solución</b></summary>

El subconjunto cambió y el DataFrame original no, o quizá Pandas emitió una advertencia (`SettingWithCopyWarning`). El problema es el mismo de las vistas y copias del Módulo 2: `df[df["dia"] == "Lunes"]` puede devolver una copia o una vista, y Pandas no garantiza cuál, de modo que asignarle un valor tiene un efecto impredecible. A veces modifica el original, a veces solo una copia que se descarta, y casi nunca hace lo que esperabas.

La solución es usar `loc` para seleccionar y asignar **en una sola operación**, sin encadenar:

```
df.loc[df["dia"] == "Lunes", "hora"] = "MODIFICADO"
```

Así Pandas sabe exactamente qué celdas modificar y lo hace sobre el DataFrame original, sin ambigüedad. La regla profesional: **nunca asignes sobre el resultado de un filtro encadenado; usa siempre `df.loc[condición, columna] = valor`**. Es el error silencioso más famoso de Pandas, y su advertencia es fácil de ignorar hasta que produce un resultado equivocado.
</details>

## 8. Tipos y conversión · Types and conversion

### 8.1 El problema: todo es texto

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)

# Intentar calcular la media de la espera falla, porque es texto
print("Tipo de la columna de espera:", df["espera_min"].dtype)
try:
    df["espera_min"].mean()
except TypeError as fallo:
    print("La media falla:", str(fallo)[:70])
````

La columna de espera es texto, y sobre texto no se puede calcular. Además, algunos valores usan coma decimal (`"42,5"`) y otros están vacíos. Convertir no es una sola operación: es un pequeño flujo.

### 8.2 Convertir de texto a número

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)

# Paso 1: unificar el separador decimal, cambiando coma por punto
espera_texto = df["espera_min"].str.replace(",", ".", regex=False)

# Paso 2: convertir a número. errors="coerce" pone NaN donde no puede convertir
df["espera_num"] = pd.to_numeric(espera_texto, errors="coerce")

print("Tipo de la nueva columna:", df["espera_num"].dtype)
print("Ahora sí se puede calcular la media:", round(df["espera_num"].mean(), 1), "min")
print("Valores que no se pudieron convertir (NaN):", df["espera_num"].isna().sum())
````

`pd.to_numeric` con `errors="coerce"` es la herramienta clave: convierte lo que puede y pone `NaN` (el marcador de faltante) donde no puede, en lugar de fallar. Así, las celdas vacías del registro se vuelven `NaN` de forma controlada, y el análisis puede continuar decidiendo qué hacer con ellas, que es asunto de la sección 9.

⚠️ **Error frecuente.** Usar `errors="coerce"` sin mirar cuántos valores se volvieron `NaN`. Si conviertes una columna y de golpe la mitad son faltantes, no tienes una columna limpia: tienes un problema de datos que la conversión silenciosa acaba de ocultar. ### 9.1 Detectar y contar los faltantes

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)
df["espera_num"] = pd.to_numeric(df["espera_min"].str.replace(",", ".", regex=False), errors="coerce")

# isna() marca cada celda faltante; sumar cuenta por columna
print("Faltantes por columna:")
print(df.isna().sum())

print(f"\nEspera: {df['espera_num'].isna().sum()} faltantes de {len(df)} "
      f"({df['espera_num'].isna().mean() * 100:.1f} %)")
````

`isna()` produce un DataFrame de verdaderos y falsos, y `sum()` los cuenta. La combinación `isna().mean()` da directamente la **proporción** de faltantes, porque la media de una columna booleana es su proporción de verdaderos, exactamente el modismo del Módulo 2.

### 9.2 Eliminar frente a imputar

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)
df["espera_num"] = pd.to_numeric(df["espera_min"].str.replace(",", ".", regex=False), errors="coerce")

n_original = len(df)

# Opción A: eliminar las filas con espera faltante
sin_faltantes = df.dropna(subset=["espera_num"])
print(f"dropna: quedan {len(sin_faltantes)} de {n_original} filas")
print(f"  Media sobre lo que queda: {sin_faltantes['espera_num'].mean():.1f} min")

# Opción B: imputar con la mediana (más robusta que la media ante la cola)
mediana = df["espera_num"].median()
imputado = df["espera_num"].fillna(mediana)
print(f"\nfillna con la mediana ({mediana:.1f}): se conservan las {n_original} filas")
print(f"  Media tras imputar: {imputado.mean():.1f} min")
````

Las dos opciones dan medias distintas, y ninguna es "la correcta" en abstracto. `dropna` elimina las filas con faltante, lo que es limpio pero descarta información y puede sesgar si los faltantes no son aleatorios. `fillna` imputa un valor, lo que conserva las filas pero inventa datos, y la media resultante depende de qué valor se imputó.

⚠️ **Error frecuente y grave.** Imputar con la media y luego reportar esa media como hallazgo. Si rellenas los faltantes con la media y después calculas la media, obtienes casi el mismo número, pero ahora calculado en parte sobre datos inventados, con una falsa sensación de completitud. La imputación es una decisión que se declara, no un truco para que no haya huecos.

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)

n_antes = len(df)

# drop_duplicates conserva la primera aparición de cada fila repetida
df_limpio = df.drop_duplicates()
print(f"drop_duplicates: {n_antes} → {len(df_limpio)} filas ({n_antes - len(df_limpio)} eliminadas)")

# Eliminar por el identificador, no por la fila completa, es distinto
df_por_id = df.drop_duplicates(subset=["id_cita"])
print(f"drop_duplicates por id: {n_antes} → {len(df_por_id)} filas")
````

### 11.2 Lo que una fecha permite preguntar

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)
df["fecha_dt"] = pd.to_datetime(df["fecha"], dayfirst=True, errors="coerce")

# El accesor .dt extrae componentes de una columna de fechas
df["dia_semana"] = df["fecha_dt"].dt.day_name()
df["dia_mes"] = df["fecha_dt"].dt.day
df["semana"] = df["fecha_dt"].dt.isocalendar().week

print("Componentes extraídos de la fecha:")
print(df[["fecha_dt", "dia_semana", "dia_mes", "semana"]].head())

print("\nCitas por día del mes (primeros cinco):")
print(df["dia_mes"].value_counts().sort_index().head())
````

Convertir texto en fecha no es cosmética: **desbloquea preguntas nuevas**. Con la fecha como texto no podías agrupar por semana, ni ordenar cronológicamente, ni calcular cuántos días pasaron entre dos citas. El accesor `.dt` extrae el día de la semana, el mes, la semana del año y decenas de componentes más, cada uno una dimensión nueva de análisis.

🔬 **Nota técnica.** Pandas almacena las fechas internamente como números enteros de nanosegundos desde una fecha de referencia (el 1 de enero de 1970), lo que permite operaciones aritméticas exactas: restar dos fechas da un intervalo, sumar un intervalo a una fecha da otra fecha. Ese diseño hace que agrupar por periodo, remuestrear una serie temporal o calcular duraciones sea directo. La conversión de texto a este tipo es lo que separa una columna de fechas usable de una lista de cadenas que solo se pueden comparar como texto.

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)
df["espera_num"] = pd.to_numeric(df["espera_min"].str.replace(",", ".", regex=False), errors="coerce")
df["especialidad"] = df["especialidad"].apply(limpiar_especialidad)

# groupby divide por especialidad, aplica una agregación y combina el resultado
media_por_esp = df.groupby("especialidad")["espera_num"].mean()
print("Espera media por especialidad:")
print(media_por_esp.sort_values(ascending=False).round(1))
````

Todo el análisis por grupos del Módulo 2, que costaba dos acumuladores y un bucle, cabe ahora en una línea: `df.groupby("especialidad")["espera_num"].mean()`. Pandas **divide** el DataFrame en grupos por especialidad, **aplica** la media a cada grupo, y **combina** los resultados en una Series.

🌱 **Analogía.** Es repartir un mazo de cartas por palo, contar cada montón y anotar los cuatro totales. `groupby` hace el reparto, la agregación hace el conteo, y Pandas junta los resultados. Tú solo declaras por qué agrupar y qué calcular.

### 12.2 Varias agregaciones a la vez

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)
df["espera_num"] = pd.to_numeric(df["espera_min"].str.replace(",", ".", regex=False), errors="coerce")
df["especialidad"] = df["especialidad"].apply(limpiar_especialidad)

# .agg aplica varias funciones de golpe y las devuelve en una tabla
resumen = df.groupby("especialidad")["espera_num"].agg(["count", "mean", "median", "max"])
print("Resumen por especialidad:")
print(resumen.round(1).sort_values("mean", ascending=False))
````

`.agg(["count", "mean", "median", "max"])` produce de una vez una tabla con cuatro estadísticos por grupo. ### 13.2 Márgenes y varias agregaciones

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)
df["espera_num"] = pd.to_numeric(df["espera_min"].str.replace(",", ".", regex=False), errors="coerce")
df["especialidad"] = df["especialidad"].apply(limpiar_especialidad)

# margins=True añade una fila y columna de totales; el conteo revela el tamaño de cada grupo
conteo = pd.pivot_table(df, values="espera_num", index="especialidad",
                        columns="dia", aggfunc="count", margins=True, margins_name="TOTAL")
print("Número de citas por especialidad y día, con totales:")
print(conteo.to_string())
````

`margins=True` añade los totales por fila y por columna, que es lo primero que mira quien lee una tabla. Y usar `count` como agregación revela cuántas citas sostienen cada celda, la comprobación de fiabilidad que ya es un reflejo a estas alturas del curso.

🔬 **Nota técnica.** `pivot_table` y `groupby` no son rivales sino dos vistas de la misma operación: `pivot_table` es esencialmente un `groupby` por dos claves seguido de un `unstack` que despliega una de ellas a las columnas. Elegir entre uno y otro es una cuestión de forma de salida, no de cálculo: `groupby` deja el resultado en formato largo, una fila por combinación, ideal para seguir procesando; `pivot_table` lo deja en formato ancho, una matriz, ideal para leer. Saber pasar de uno a otro con `stack` y `unstack` es una de las destrezas más útiles de Pandas.

### 14.1 Apilar con concat

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)

# concat apila DataFrames: útil para juntar el registro de varios meses
enero = df.iloc[:300].copy()
febrero = df.iloc[300:600].copy()

trimestre = pd.concat([enero, febrero], ignore_index=True)
print("Enero:", len(enero), "· Febrero:", len(febrero), "· Juntos:", len(trimestre))
print("¿La suma cuadra?", len(trimestre) == len(enero) + len(febrero))
````

### 14.2 Cruzar con merge

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)
df["especialidad"] = df["especialidad"].apply(limpiar_especialidad)

# Una segunda tabla: información de cada especialidad (un catálogo aparte)
catalogo = pd.DataFrame({
    "especialidad": ["Medicina interna", "Pediatría", "Ortopedia", "Ginecología"],
    "piso": [2, 3, 1, 4],
    "consultorios": [5, 4, 3, 3],
})

# merge cruza las dos tablas por la columna común
cruzado = df.merge(catalogo, on="especialidad", how="left")
print("Columnas tras el cruce:", list(cruzado.columns))
print("\nEjemplo: cada cita ahora sabe en qué piso se atiende:")
print(cruzado[["id_cita", "especialidad", "piso", "consultorios"]].head(3))
print("\n¿Se conservaron todas las filas?", len(cruzado) == len(df))
````

`merge` es el cruce de tablas por una clave común, la operación que en bases de datos se llama *join*. Cada cita, que solo tenía su especialidad, ahora sabe también en qué piso y con cuántos consultorios se atiende, porque esa información vivía en otra tabla y `merge` la trajo.

### 14.3 El tipo de unión decide qué se conserva

````python
import pandas as pd
df = pd.DataFrame(filas, columns=COLUMNAS)
df["especialidad"] = df["especialidad"].apply(limpiar_especialidad)

# Un catálogo INCOMPLETO: le falta Ginecología
catalogo_incompleto = pd.DataFrame({
    "especialidad": ["Medicina interna", "Pediatría", "Ortopedia"],
    "piso": [2, 3, 1],
})

# how="inner" conserva solo las filas con coincidencia en AMBAS tablas
inner = df.merge(catalogo_incompleto, on="especialidad", how="inner")
# how="left" conserva TODAS las filas de la izquierda, con NaN donde no hay coincidencia
left = df.merge(catalogo_incompleto, on="especialidad", how="left")

print("Filas originales      :", len(df))
print("Tras inner join       :", len(inner), "· se perdieron las de Ginecología")
print("Tras left join        :", len(left), "· se conservan, con piso vacío")
print("Pisos faltantes en left:", left["piso"].isna().sum())
````

🕵️ Este es el error silencioso más costoso de las uniones. `how="inner"` conserva solo las filas que casan en ambas tablas, de modo que si el catálogo no tiene Ginecología, **todas las citas de Ginecología desaparecen del resultado sin ningún error**. `how="left"` conserva todas las filas de la tabla izquierda y pone `NaN` donde no hay coincidencia, haciendo visible el problema en lugar de ocultarlo.

⚠️ **Regla profesional.** Después de todo `merge`, comprueba el número de filas. Si esperabas conservar todas y tienes menos, un `inner join` te borró las que no casaron; si tienes más, la clave no era única y las filas se multiplicaron. Es el mismo `zip` que recortaba del Módulo 2, con otro disfraz: cuando la correspondencia entre datos depende de una clave, verifica siempre que la unión hizo lo que creías.

### 🎯 Ejercicio rápido 1

¿Qué diferencia hay entre un `inner join` y un `left join`, y por qué el `inner` puede perder datos en silencio?

<details><summary><b>Ver solución</b></summary>

Un `inner join` conserva solo las filas cuya clave aparece en ambas tablas, mientras que un `left join` conserva todas las filas de la tabla izquierda y rellena con NaN las columnas de la derecha cuando no hay coincidencia. El `inner` pierde datos en silencio porque, si a la tabla de la derecha le falta alguna clave, todas las filas de la izquierda con esa clave desaparecen del resultado sin producir ningún error ni advertencia. En el caso de la IPS, un catálogo al que le falte una especialidad borraría todas las citas de esa especialidad del análisis. Por eso conviene preferir `left join` cuando la tabla de la izquierda es la que debe conservarse íntegra, y comprobar siempre el número de filas después de unir.
</details>

### 15.2 Barras desde una agregación

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(filas, columns=COLUMNAS)
df["espera_num"] = pd.to_numeric(df["espera_min"].str.replace(",", ".", regex=False), errors="coerce")
df["especialidad"] = df["especialidad"].apply(limpiar_especialidad)

# groupby produce la agregación; .plot.bar la dibuja, encadenados
medias = df.groupby("especialidad")["espera_num"].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(9, 4.5))
medias.plot.bar(color="#07a3b2", ax=ax)
ax.axhline(30, color="#c1121f", linestyle="--", label="Meta: 30 min")
ax.set_title("Espera media por especialidad")
ax.set_ylabel("Minutos")
ax.set_ylim(0, medias.max() * 1.15)      # eje desde cero, sin excepción
ax.tick_params(axis="x", labelrotation=20)
ax.legend()
plt.tight_layout()
plt.show()
````

Análisis y gráfico en una tubería: `df.groupby(...).mean().sort_values().plot.bar()`. La agregación fluye directa al dibujo. Y las reglas de honestidad del Módulo 2 siguen vigentes: el eje vertical empieza en cero, y la línea de la meta da un punto de referencia para interpretar las alturas.

### 15.3 Líneas para la evolución horaria

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(filas, columns=COLUMNAS)
df["espera_num"] = pd.to_numeric(df["espera_min"].str.replace(",", ".", regex=False), errors="coerce")
df["especialidad"] = df["especialidad"].apply(limpiar_especialidad)
df["hora"] = pd.to_numeric(df["hora"])

# Una tabla dinámica alimenta directamente un gráfico de líneas
tabla = pd.pivot_table(df, values="espera_num", index="hora", columns="especialidad", aggfunc="mean")

fig, ax = plt.subplots(figsize=(9.5, 4.5))
tabla.plot(ax=ax, marker="o")
ax.axhline(30, color="#c1121f", linestyle=":", label="Meta")
ax.set_title("La espera crece a lo largo de la jornada")
ax.set_xlabel("Hora de la cita")
ax.set_ylabel("Espera media (min)")
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
````

### 17.2 El protocolo, paso a paso

El protocolo del dataset de accidentes de Nueva York define cuatro reglas de imputación, que aquí traducimos a nuestras columnas equivalentes:

| Regla del protocolo NY | Nuestra columna equivalente | Acción |
|---|---|---|
| Imputar `ZIP CODE` por la moda dentro de cada municipio | Especialidad por la moda dentro de cada día | Rellenar con el valor más frecuente del grupo |
| Imputar coordenadas por la media | Espera por la media dentro de cada especialidad | Rellenar con el promedio del grupo |
| Categóricas faltantes con `Unspecified` o `Unknown` | Categóricas faltantes con `"Sin dato"` | Marcar explícitamente lo que falta |
| Convertir `DATE` a `datetime` | Convertir `fecha` a `datetime` | Parsear el texto a fecha |

````python
import pandas as pd

# Partimos del registro crudo y aplicamos el protocolo completo, paso a paso
df = pd.DataFrame(filas, columns=COLUMNAS)
print("PROTOCOLO DE LIMPIEZA ESTANDARIZADO")
print("=" * 60)
print(f"Paso 0 · Registro crudo: {df.shape[0]} filas, {df.shape[1]} columnas")

# Paso 1: eliminar duplicados completos
df = df.drop_duplicates().reset_index(drop=True)
print(f"Paso 1 · Sin duplicados: {df.shape[0]} filas")

# Paso 2: normalizar la categoría de texto
df["especialidad"] = df["especialidad"].apply(limpiar_especialidad)
print(f"Paso 2 · Especialidad normalizada: {df['especialidad'].nunique()} categorías")

# Paso 3: convertir tipos (espera a número, hora a entero)
df["espera_min"] = pd.to_numeric(df["espera_min"].str.replace(",", ".", regex=False), errors="coerce")
df["hora"] = pd.to_numeric(df["hora"], errors="coerce")
print(f"Paso 3 · Tipos convertidos · esperas faltantes: {df['espera_min'].isna().sum()}")

# Paso 4: imputar la espera faltante por la MEDIA de su especialidad (regla de coordenadas)
df["espera_min"] = df.groupby("especialidad")["espera_min"].transform(lambda s: s.fillna(s.mean()))
print(f"Paso 4 · Espera imputada por media de grupo · faltantes: {df['espera_min'].isna().sum()}")

# Paso 5: convertir la fecha a datetime (regla DATE)
df["fecha"] = pd.to_datetime(df["fecha"], dayfirst=True, errors="coerce")
print(f"Paso 5 · Fecha convertida: {df['fecha'].dtype}")

print("=" * 60)
print(f"Resultado: {df.shape[0]} filas limpias, listas para analizar")
````

Fíjate en `transform` del paso 4: aplica una operación por grupo pero **devuelve el resultado alineado con las filas originales**, de modo que cada faltante se rellena con la media de su propia especialidad, no con la media global. Es la forma correcta de imputar por grupo, y una función de Pandas que vale la pena conocer.

🔬 **Nota técnica.** El orden del protocolo no es arbitrario. Los duplicados se eliminan primero, para no sesgar las medias con filas repetidas. La normalización de la categoría va antes de la imputación por grupo, porque si la especialidad estuviera sucia, los grupos estarían fragmentados y la media de cada uno sería errónea. La conversión de tipos precede a la imputación numérica, porque no se puede promediar texto. Cada paso habilita al siguiente, y alterar el orden produce un resultado distinto y peor. Documentar ese orden es parte del protocolo, no un detalle.
