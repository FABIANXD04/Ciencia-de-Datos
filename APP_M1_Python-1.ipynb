### 1.1 Qué significa programar

Programar es escribir instrucciones precisas para que una máquina las ejecute siempre igual. Nada más, y nada menos. La dificultad no está en la máquina, que es obediente hasta la exasperación, sino en la precisión: casi todo lo que damos por obvio al hablar con una persona hay que decirlo de forma explícita.

Ejecuta la celda siguiente. Es un programa completo.

````python
print("Hola, ya estoy programando en Python")
````

Ya está. Eso fue programar.

`print` es una **función**: un procedimiento que ya existe en Python y que hace una cosa concreta, mostrar en pantalla lo que se le entregue entre paréntesis. Las comillas indican que lo de dentro es texto literal, que debe tomarse tal como está escrito.

🌱 **Analogía.** Piensa en una receta escrita para alguien que sigue instrucciones al pie de la letra y no interpreta nada. Si la receta dice "añade sal", esa persona preguntará cuánta. Programar es escribir la receta con ese nivel de detalle: no porque la máquina sea torpe, sino porque no supone nada.

🔬 **Nota técnica.** `print` es una función integrada (*built-in*) del espacio de nombres `builtins`, disponible sin importar nada. Su firma es `print(*objects, sep=' ', end='\n', file=None, flush=False)`. Los parámetros `sep` y `end` explican por qué varios argumentos aparecen separados por espacio y por qué cada llamada termina en salto de línea; ambos son modificables.

Ahora una instrucción con un dato del caso.

````python
print("Citas programadas hoy en consulta externa:", 240)
````

Aparecieron dos cosas nuevas. La **coma** separa varios elementos dentro del mismo `print`, y Python los muestra uno tras otro con un espacio entre ellos. Y el número `240` va **sin comillas**, porque no es texto: es un número con el que se puede calcular.

Esa distinción entre texto y número parece menor y es la fuente del error más común de las primeras horas. La retomaremos en la sección 3.

### 1.4 Los errores no son fracasos

La celda siguiente está mal escrita **a propósito**. Ejecútala y no te alarmes por el bloque rojo.

````python
print(Hola)
````

Apareció un bloque rojo largo y de aspecto intimidante. Se lee así: **la última línea es la que importa**. Todo lo anterior es el recorrido interno de Python hasta el punto del fallo, útil en programas grandes e irrelevante ahora.

La última línea dice `NameError: name 'Hola' is not defined`. Traducido: sin comillas, Python interpreta `Hola` como el **nombre de un dato guardado previamente**, lo busca en su memoria, no lo encuentra y avisa. No es un regaño ni una falla del programa: es un diagnóstico bastante preciso.

Corrígelo ejecutando la celda siguiente.

````python
print("Hola")     # con comillas, Python lo entiende como texto y funciona
````

### ⚠️ Errores frecuentes al empezar

### 1.5 El cuaderno tiene memoria

Un cuaderno no es un archivo de texto que se lee de arriba abajo: es una sesión viva. Detrás hay un proceso llamado **núcleo** (*kernel*) que ejecuta el código y **recuerda** todo lo que ha creado. Al conjunto de datos vivos en esa memoria se le llama **estado** (*state*).

````python
# Guardamos un dato con un nombre. Esto se llama crear una variable.
citas_programadas = 240
print(citas_programadas)
````

En una celda distinta, sin volver a escribirlo, el dato sigue disponible.

````python
citas_atendidas = 128
citas_pendientes = citas_programadas - citas_atendidas    # usa la variable de la celda anterior
print("Pendientes:", citas_pendientes)
````

### 1.6 El orden de ejecución: la trampa más común

Ejecuta la celda siguiente **tres veces seguidas** y observa el número que imprime cada vez.

````python
citas_pendientes = citas_pendientes - 10    # le restamos 10 al valor que ya había
print(citas_pendientes)
````

El código es idéntico las tres veces y el resultado cambia. La razón es que la celda no calcula un valor desde cero: **modifica el que ya estaba en memoria**. Cada ejecución parte del estado que dejó la anterior.

🧩 **Trazado de estado.** Sigue el valor de `citas_pendientes` a lo largo de las ejecuciones. Esta tabla es el hábito mental que conviene adquirir: antes de ejecutar, pregúntate en qué estado está la sesión.

| Momento | `citas_programadas` | `citas_atendidas` | `citas_pendientes` |
|---|---|---|---|
| Tras la celda de 1.5 | 240 | sin definir | sin definir |
| Tras la primera celda de 1.5 (segunda parte) | 240 | 128 | 112 |
| Tras la primera ejecución de 1.6 | 240 | 128 | 102 |
| Tras la segunda ejecución | 240 | 128 | 92 |
| Tras la tercera ejecución | 240 | 128 | 82 |

Esto tiene una consecuencia profesional directa. Un cuaderno cuyos resultados dependen del orden en que se pulsaron las celdas **no es evidencia verificable**, por correctas que sean sus conclusiones. Por eso la regla de oro de esta unidad, que aparecerá en todas las listas de verificación: antes de entregar cualquier cuaderno, **reinicia el núcleo y ejecuta todo de arriba abajo**. En Colab, `Entorno de ejecución → Reiniciar y ejecutar todo`.

La forma de evitar el problema es escribir celdas que no dependan de cuántas veces se ejecutaron: partir siempre de una constante que no se modifica.

````python
CITAS_PROGRAMADAS = 240        # constante: se escribe en mayúsculas y no se modifica nunca
citas_atendidas = 128

citas_pendientes = CITAS_PROGRAMADAS - citas_atendidas   # se recalcula desde el origen
print("Pendientes:", citas_pendientes)
print("Ejecuta esta celda las veces que quieras: el resultado no cambia.")
````

El nombre en MAYÚSCULAS es una convención: señala a quien lea el código que ese valor es una **constante** y no debe modificarse. Python no lo impide, pero la comunidad lo respeta y las herramientas de revisión lo verifican.

🔬 **Nota técnica.** Python no tiene constantes reales: `CITAS_PROGRAMADAS` es una variable ordinaria y `CITAS_PROGRAMADAS = 1` la reasignaría sin protestar. La convención está recogida en PEP 8, que reserva `MAYUSCULAS_CON_GUION_BAJO` para valores destinados a no cambiar durante la vida del módulo. Lo que sí es real es la ausencia de tipado estático: el nombre es una referencia y puede apuntar a cualquier objeto.

### 2.1 Por qué el módulo abre con una metodología y no con código

CRISP-DM (*Cross Industry Standard Process for Data Mining*) es el marco de referencia más extendido para estructurar proyectos de datos. Se publicó en 2000 y sigue vigente por una razón que conviene entender: no describe técnicas, describe **decisiones**, y las decisiones envejecen mucho más despacio que los algoritmos (Chapman et al., 2000).

| Fase | Pregunta que responde | Qué produce |
|---|---|---|
| 1 · Comprensión del negocio | ¿Qué decisión queremos mejorar? | Objetivo de negocio, objetivo analítico, criterio de éxito |
| 2 · Comprensión de los datos | ¿Con qué datos contamos y qué calidad tienen? | Inventario de fuentes, exploración inicial, hallazgos de calidad |
| 3 · Preparación de los datos | ¿Cómo dejamos los datos utilizables? | Conjunto limpio, transformado y documentado |
| 4 · Modelado | ¿Qué técnica responde la pregunta? | Modelos ajustados y parámetros |
| 5 · Evaluación | ¿El resultado sirve para la decisión inicial? | Métricas, comparación con la línea base, decisión de continuar |
| 6 · Despliegue | ¿Cómo llega esto a quien decide? | Producto en uso, monitoreo, documentación |

Dos rasgos del ciclo suelen pasarse por alto y son los que más ahorran trabajo. El primero es que **es iterativo**: se vuelve de la fase 4 a la 3, y de la 5 a la 1, tantas veces como haga falta. El segundo es que la fase 5 incluye de forma explícita la posibilidad de **detener el proyecto**, algo que rara vez se contempla y que evita sostener durante meses una solución que no mejora ninguna decisión.

🌱 **Analogía.** Antes de construir una casa alguien decide para quién es, cuántas personas la habitarán y con qué presupuesto. Empezar por los planos eléctricos es posible, y también es la forma más segura de terminar con una instalación impecable en una casa que nadie necesitaba.

🔬 **Nota técnica.** CRISP-DM no es la única opción. En equipos con producto en producción conviene compararlo con marcos que sí cubren el ciclo de vida operativo, como TDSP de Microsoft o las prácticas de MLOps, que formalizan monitoreo, reentrenamiento y control de versiones de datos y modelos. CRISP-DM es deliberadamente agnóstico respecto de la infraestructura, lo que es su fortaleza como marco de razonamiento y su límite como marco de operación.

### 2.2 La distinción que más cuesta y más ahorra

El **objetivo de negocio** nombra un resultado deseado en el lenguaje de la organización. El **objetivo analítico** nombra una cantidad que se puede estimar a partir de datos. Confundirlos produce modelos que optimizan algo que a nadie le sirve.

| | Enunciado en el caso de la IPS |
|---|---|
| Objetivo de negocio | Reducir la espera de los pacientes de consulta externa en el próximo trimestre |
| Objetivo analítico | Estimar el tiempo de espera, en minutos, de cada cita según su hora de programación, su especialidad y el día de la semana |

Fíjate en tres rasgos del objetivo analítico que el de negocio no tiene: nombra una **cantidad medible** (minutos), declara la **unidad de análisis** (cada cita) y enuncia **con qué información** se pretende estimarla. Sin esos tres elementos no hay proyecto de datos, hay una aspiración.

### 2.3 El acta mínima, escrita con variables

Un acta de proyecto es el documento donde se fija todo lo anterior antes de tocar un solo dato. Aquí la escribiremos con las herramientas que ya conoces: variables y `print`.

````python
# ---------------------------------------------------------------------
# Acta de idea de proyecto · versión mínima
# Cada dato del acta se guarda en una variable con nombre explícito.
# ---------------------------------------------------------------------
ORGANIZACION       = "IPS Salud Meridiano"
SECTOR             = "Salud · consulta externa"
OBJETIVO_NEGOCIO   = "Reducir la espera de los pacientes de consulta externa en el próximo trimestre"
OBJETIVO_ANALITICO = "Estimar el tiempo de espera en minutos de cada cita según hora, especialidad y día"
UNIDAD_DE_ANALISIS = "Una cita de consulta externa"
HORIZONTE_DIAS     = 90

LINEA_BASE_MIN = 42.0      # espera promedio que reporta hoy el sistema de agendamiento
META_MIN       = 30.0      # espera promedio que la gerencia considera aceptable

reduccion_absoluta = LINEA_BASE_MIN - META_MIN
reduccion_relativa = reduccion_absoluta / LINEA_BASE_MIN * 100

print("ACTA DE IDEA DE PROYECTO")
print("-" * 62)
print("Organización       :", ORGANIZACION)
print("Sector             :", SECTOR)
print("Unidad de análisis :", UNIDAD_DE_ANALISIS)
print("Horizonte (días)   :", HORIZONTE_DIAS)
print("-" * 62)
print("Objetivo de negocio  :", OBJETIVO_NEGOCIO)
print("Objetivo analítico   :", OBJETIVO_ANALITICO)
print("-" * 62)
print("Línea base (min)   :", LINEA_BASE_MIN)
print("Meta (min)         :", META_MIN)
print("Reducción buscada  :", reduccion_absoluta, "minutos, equivalentes a")
print("                    ", reduccion_relativa, "por ciento")
````

Ese `reduccion_relativa` salió con demasiados decimales. Es correcto pero ilegible, y el formato de salida no es un detalle cosmético cuando alguien va a decidir con esa cifra. En la sección 3.7 aprenderás a controlarlo.

⚠️ **Error frecuente.** Fijar la meta antes que la línea base. Sin saber cuánto se espera hoy, "reducir a 30 minutos" puede ser trivial o imposible, y no hay forma de saber cuál de las dos cosas es. La línea base (*baseline*) es la primera cifra que debe producir cualquier proyecto de datos, y con frecuencia es también la que cambia la conversación: no es raro que medir el punto de partida vuelva innecesario el modelo.

### 3.4 Conversión de tipos

Cuando un dato llega como texto y necesitas calcular con él, hay que convertirlo de forma explícita.

````python
texto_citas = "240"          # llegó como texto, por ejemplo desde un archivo
print(type(texto_citas))

numero_citas = int(texto_citas)      # int() convierte texto a entero
print(type(numero_citas))
print(numero_citas + 8)              # ahora sí se puede calcular

espera_texto = "42.5"
espera_num = float(espera_texto)     # float() convierte texto a decimal
print(espera_num * 2)

print("Se atendieron " + str(numero_citas) + " citas")   # str() convierte número a texto
````

🔬 **Nota técnica.** `int("42.5")` lanza `ValueError`: la conversión directa de texto a entero solo admite representaciones enteras. La ruta correcta es `int(float("42.5"))`, que trunca hacia cero, o `round(float("42.5"))`, que redondea al par más próximo por la regla del redondeo bancario. Esa diferencia entre truncar y redondear produce discrepancias de conteo en informes de volumen, y es una de las causas más comunes de que dos áreas de la misma organización reporten cifras distintas.

### 3.5 Operadores aritméticos

| Operador | Operación | Ejemplo | Resultado |
|---|---|---|---|
| `+` | Suma | `240 + 8` | `248` |
| `-` | Resta | `240 - 128` | `112` |
| `*` | Multiplicación | `20 * 3` | `60` |
| `/` | División | `240 / 7` | `34.2857…` |
| `//` | División entera | `240 // 7` | `34` |
| `%` | Residuo (módulo) | `240 % 7` | `2` |
| `**` | Potencia | `2 ** 3` | `8` |

El indicador central del caso es el **tiempo medio de espera**, que se define como

$$\bar{t} = \frac{1}{n}\sum_{i=1}^{n} t_i$$

donde $t_i$ es el tiempo de espera de la cita $i$ y $n$ es el número total de citas atendidas. Es la media aritmética de toda la vida, escrita en la notación que encontrarás en la literatura del área.

````python
# Datos agregados de una jornada, tomados del sistema de agendamiento.
CITAS_ATENDIDAS       = 128
ESPERA_TOTAL_MIN      = 5376        # suma de los minutos de espera de todas las citas
DURACION_CITA_MIN     = 20
JORNADA_MIN           = 10 * 60     # de 7:00 a 17:00

espera_media = ESPERA_TOTAL_MIN / CITAS_ATENDIDAS
capacidad_teorica = JORNADA_MIN // DURACION_CITA_MIN        # cuántas citas caben por consultorio
consultorios_necesarios = CITAS_ATENDIDAS / capacidad_teorica
citas_sobrantes = CITAS_ATENDIDAS % capacidad_teorica       # las que no completan un consultorio

print("Espera media (min)          :", espera_media)
print("Capacidad por consultorio   :", capacidad_teorica)
print("Consultorios necesarios     :", consultorios_necesarios)
print("Citas que sobran del último :", citas_sobrantes)
````

⚠️ **Error frecuente.** Confundir `/` con `//`. La primera devuelve siempre un decimal, incluso cuando la división es exacta: `10 / 5` da `2.0`, no `2`. La segunda descarta la parte decimal. Usar una por otra al contar unidades indivisibles, como consultorios o personas, produce resultados que no significan nada: cuatro coma dos consultorios no existen.

### 3.6 Operadores de comparación

Comparar produce siempre un valor lógico, `True` o `False`. Son la materia prima de las decisiones que escribirás en la sección 4.

| Operador | Significado |
|---|---|
| `==` | Igual a |
| `!=` | Distinto de |
| `>` `<` | Mayor que, menor que |
| `>=` `<=` | Mayor o igual, menor o igual |

````python
espera_media = 42.0
META_MIN = 30.0

print("¿Se cumple la meta?          ", espera_media <= META_MIN)
print("¿Excede la meta?             ", espera_media > META_MIN)
print("¿Excede en más de 10 minutos?", espera_media - META_MIN > 10)

# Los operadores lógicos combinan condiciones.
agenda_saturada = True
es_hora_pico = False

print("Saturada Y hora pico :", agenda_saturada and es_hora_pico)
print("Saturada O hora pico :", agenda_saturada or es_hora_pico)
print("NO saturada          :", not agenda_saturada)
````

⚠️ **Error frecuente.** Escribir `=` cuando se quiere comparar. Un solo signo igual **asigna**; dos signos iguales **comparan**. `espera = 30` guarda el valor; `espera == 30` pregunta si vale treinta. Python detecta el caso más burdo con un `SyntaxError`, pero no todos.

### 3.7 Formato de salida: que la cifra se pueda leer

Las **f-strings** son la forma moderna de construir texto con valores dentro. Se escribe una `f` antes de la comilla y las variables van entre llaves.

````python
especialidad = "Medicina interna"
espera_media = 42.0
META_MIN = 30.0
exceso_pct = (espera_media - META_MIN) / META_MIN * 100

# Sin formato: correcto pero ilegible.
print("Exceso sobre la meta:", exceso_pct)

# Con f-string y formato: :.1f muestra un decimal, :>6 alinea a la derecha en 6 espacios.
print(f"Especialidad         : {especialidad}")
print(f"Espera media         : {espera_media:>6.1f} min")
print(f"Meta                 : {META_MIN:>6.1f} min")
print(f"Exceso sobre la meta : {exceso_pct:>6.1f} %")
````

Los dos puntos dentro de las llaves abren la **especificación de formato**. Los tres que usarás casi siempre son `:.1f` para fijar decimales, `:>8` para alinear a la derecha en un ancho dado y `:,` para separar los miles.

🔬 **Nota técnica.** Las f-strings, introducidas en PEP 498, se compilan a concatenación eficiente en tiempo de compilación y son notablemente más rápidas que `%` o `str.format`. Admiten expresiones completas dentro de las llaves, y el sufijo `=` es un recurso de depuración poco conocido y muy útil: `print(f"{espera_media=}")` imprime a la vez el nombre y el valor.
